### Why this week changes everything

Last week you tested the rock-paper-scissors judge by editing two variables and re-running the
cell nine times. This week you will test all nine combinations in **nine lines**, in one run,

That is what a function buys you: a piece of logic with a **name**, that takes **inputs** and
hands back a **result**, so you can use it a hundred times without retyping it.

> **`print` is not `return`.** `print` shows a human something. `return` hands a value back to
