enough, if you use a **triple-quoted string** — three quote marks let one string run over as
many lines as you like:

```python
print("""
first line
second line
""")
```

Two things will bite you here, and both are worth meeting now rather than in week 9:

- **Backslashes.** The 3-D fonts (`Isometric1`, `Larry 3D`, `Alpha`) are drawn mostly out of
  `\`, and Python reads a backslash as the *start of an escape code* — `\t` silently becomes
  a tab and your art melts. Put an `r` immediately before the opening quotes,
  `print(r"""...""")`, and Python prints the characters exactly as you pasted them. `Modular`
  above contains no backslashes at all, which is what makes it the safe first pick.
- **Spacing.** The leading spaces on each line *are* the picture. Paste the art in untouched
  — do not tidy up the indentation to line it up with your code.

Nothing here is examinable. It is here because poking at a language is how you learn it.
