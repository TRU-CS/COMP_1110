### What a class is for

A dictionary can hold a robot: `{"name": "Clanky", "hp": 30, "power": 6}`. So why bother with a
class?

Because a class keeps the **data** and the **things you can do to it** in one place. A dictionary
of a robot cannot attack. An object can. And when your project has fifteen functions that all
take the same six-key dictionary as their first argument, you have written a class by accident —
badly.
