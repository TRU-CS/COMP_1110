### The two ideas that make strings click

1. **A string is a sequence.** `word[0]` is the first character, `word[-1]` is the last, and
   `word[2:5]` is a slice — from index 2 up to *but not including* 5.
2. **Strings are immutable.** No method ever changes a string. `text.upper()` **returns a new
   string** and leaves `text` exactly as it was. Forgetting this is the single most common
   string bug, and it is in Part C.

## 🔐 Part B — Secret Messages

The **Caesar cipher** shifts every letter along the alphabet by a fixed amount. With a shift of
3, `a` becomes `d`, `b` becomes `e`, and `z` wraps around to `c`. Julius Caesar really used it.
It is hopeless as security and perfect as a string exercise.

To shift a letter you need two built-in functions:

````python
print(ord("a"))     # the number Python stores for the letter a
print(ord("b"))
print(chr(97))      # ...and back again
print(chr(ord("a") + 3))
````

## 💅 Part C — f-strings: output that lines up

New this week, and you will use it for the rest of the course. Put an `f` before the quote and
you can drop variables straight into the text, with formatting after a colon:

````python
item = "Chocolate"
price = 2.75
quantity = 3

print(f"{item} x{quantity} = ${price * quantity:.2f}")
print(f"|{item:<15}|{price:>8.2f}|")   # <15 = left in 15 columns, >8.2f = right, 2 decimals
print(f"{'Item':<15}{'Price':>8}")
````
