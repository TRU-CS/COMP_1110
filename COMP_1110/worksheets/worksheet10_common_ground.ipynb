### Three structures, three jobs

| Structure | Ordered? | Changeable? | Duplicates? | Use it when |
| :--- | :--- | :--- | :--- | :--- |
| **list** | yes | yes | yes | a sequence you will edit |
| **tuple** | yes | **no** | yes | a fixed record: a pair, a coordinate, a row |
| **set** | no | yes | **no** | membership and overlap: "is it in there?", "what do we share?" |

A tuple is the right shape for "a name **and** a score travelling together". A set is the right
shape for "what do these two people have in common?" — and it answers that in one character.

- `only_mine(a, b)` — in `a` but not `b`
- `similarity(a, b)` — how alike two people are, as a number from 0 to 1:
  **the size of the overlap divided by the size of the union**

That last one has a name — the **Jaccard index** — and it is genuinely used in recommendation
systems. Two identical sets score `1.0`; two with nothing in common score `0.0`. Two empty sets
score... your call. Say what you decided in the docstring.
