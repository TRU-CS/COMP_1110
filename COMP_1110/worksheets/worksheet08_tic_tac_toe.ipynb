### The thing that makes lists different

Strings are **immutable** — no method changes them. Lists are **mutable** — many methods change
them *in place*, and hand you back `None`.

```python
words = ["b", "a", "c"]
words.sort()          # changes words, returns None
tidy = sorted(words)  # leaves words alone, returns a NEW list
```

Half the bugs this week are `x = x.sort()`, which throws your list away and leaves you with
`None`. The other half are aliasing, which is Part D.
