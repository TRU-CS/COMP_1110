### The three parts of every loop

Almost every bug in this chapter is one of these three going missing:

1. **Initialise** — set the variable up *before* the loop.
2. **Test** — the condition that keeps the loop going.
3. **Change** — something inside the loop must move the test towards `False`,
   or the loop runs forever.

### B2 — The smart bot 🧠

`smart_bot(secret)`: the bot knows the answer is somewhere between `low = 1` and `high = 100`.
Each turn it guesses the **middle** of that range, `(low + high) // 2`, and then:

- if the guess is too low, nothing below the guess is possible any more → move `low` up
- if the guess is too high → move `high` down
- if it is right → stop

### B3 — Play against the bot 🎮

Now the other way round: **you** think of a number and the bot interrogates you. Run the cell

This one uses real `input()`: on the practice platform the run stops at each question with a
caret in the output — type one letter, press Enter, and the bot asks the next one. Play at least
one round — then try to *cheat* by lying, and watch the bot corner itself into an impossible
range.

````python
# Read this code before running it. Then run it and think of a number 1-100.
low = 1
high = 100
guesses = 0

while True:
    guess = (low + high) // 2
    guesses = guesses + 1
    answer = input("Is it " + str(guess) + "? (h = too high, l = too low, c = correct) ")
    if answer == "c":
        print("Got it in", guesses, "guesses.")
        break
    elif answer == "h":
        high = guess - 1
    elif answer == "l":
        low = guess + 1
    else:
        print("Please answer h, l or c.")
    if low > high:
        print("You lied to me. There is no number left that fits your answers.")
        break
````
