## Part A — Strings

Quick reminders:

- Indexes start at **0**; `s[-1]` is the **last** character.
- `s[a:b]` goes from index `a` **up to but not including** `b`.
  Leave out `a` to start at the beginning, `b` to go to the end.
- Strings are **immutable**: `s[0] = "x"` is an error, and methods like
  `upper()` and `strip()` return a **new** string — they never change `s`.

### A1 — indexing a string *(easy)*

```python

```

**What does this print?**

<details>
<summary><b>Answer</b></summary>

```text
P h n
```

Index 0 is `P`, index 3 is the fourth letter `h`, and `-1` counts from the end,
so it is the last letter `n`. `print` puts one space between the three items.

</details>

## Part B — Lists

Quick reminders:

- Indexing and slicing work exactly like strings: `items[0]`, `items[-1]`,
  `items[1:3]`.
- Lists are **mutable**: `items[0] = "x"` changes the list in place.
- `append`, `insert`, `extend`, `remove`, `sort` and `reverse` change the list and
  **return `None`**. `pop` changes the list **and** returns the item it removed.
- `b = a` does **not** copy a list — both names refer to the same list.
  Use `a.copy()` or `a[:]` for a real copy.

### B1 — indexing a list *(easy)*

```python
fruits = ["apple", "banana", "cherry", "date"]
print(fruits[1])
print(fruits[-1])
print(len(fruits))
```

**What does this print?**

<details>
<summary><b>Answer</b></summary>

```text
banana
date
4
```

Index 1 is the **second** item. `-1` is the last item. `len` counts the items: 4.
Printing a single string from a list shows it **without** quotes.

</details>

### B2 — index out of range *(easy)*

```python

```

**What happens, and what is the largest index you can use here?**

<details>
<summary><b>Answer</b></summary>

It crashes with

```text

```

A list of 4 items has indexes **0 to 3**, so `fruits[3]` (or `fruits[-1]`) is the
last item. Index 4 would be a fifth item, which does not exist.

</details>


</details>

## Part C — Dictionaries

Quick reminders:

- A dictionary maps **keys** to **values**: `ages = {"Ana": 19, "Ben": 21}`.
- Look up by **key**, not by position: `ages["Ana"]`. A missing key is a
  `KeyError`.
- `d[key] = value` **adds** the key if it is new and **replaces** the value if it
  is not. Keys are unique.
- `in` checks the **keys**. `d.get(key, default)` never crashes.
- `for key in d` loops over the keys; `for key, value in d.items()` gives both.

### C1 — looking up a value *(easy)*

```python
ages = {"Ana": 19, "Ben": 21, "Cy": 20}
print(ages["Ben"])
print(len(ages))
```

**What does this print?**

<details>
<summary><b>Answer</b></summary>

```text
21
3
```

`ages["Ben"]` looks up the **key** `"Ben"` and gives its value, 21. `len` counts
the key–value pairs: 3.

</details>

```text

```

`"pear"` is a new key, so it is **added**. `"apple"` already exists, so its value
is **replaced** by `5 + 2`. A dictionary prints with braces and `key: value`
pairs.

</details>

### C3 — a missing key *(easy)*

</details>

### C11 — removing an entry *(medium)*

```python
cart = {"milk": 2, "eggs": 12, "bread": 1}
del cart["eggs"]
cart["milk"] = 0
print(cart)
print(len(cart))
```

**What does this print?**

<details>
<summary><b>Answer</b></summary>

```text

```

### C15 — list or dictionary? *(hard)*
