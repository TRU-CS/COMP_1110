## You have been using objects since Week 1

## Attributes — the data inside an object

An **attribute** is a variable that lives on an object. You reach it with a dot,
and you assign it with a dot:

```python
amara.name = "Amara"
```

Each instance carries its own attributes. Setting `amara.name` does nothing to
`liam`.

Attributes are how the object remembers things between method calls.

````python
class Student:
    """A single student enrolled in a course."""


amara = Student()
amara.name = "Amara"
amara.grades = [91, 84]

liam = Student()
liam.name = "Liam"
liam.grades = [70]

print(amara.name, amara.grades)
print(liam.name, liam.grades)
````

This works, but it is dangerous: nothing forces `liam` to *have* a `grades`
attribute. Forget one line and you get an `AttributeError` much later.
That is exactly what `__init__` fixes.

## `__init__` — the constructor

`__init__` is a method Python calls **automatically** the moment an instance is
created. Its job: set up the attributes so no instance is ever half-built.

```python
amara = Student("Amara")   # Python runs __init__(amara, "Amara")
```

- The name has two underscores on each side — a "dunder" method.
- You never call `__init__` yourself; you call the class.
- Give parameters defaults where it makes sense.

````python
class Student:
    """A single student, with a name and a list of grades."""

    def __init__(self, name):
        """Create a student with the given name and no grades yet."""
        self.name = name
        self.grades = []


amara = Student("Amara")
liam = Student("Liam")

print(amara.name, amara.grades)
print(liam.name, liam.grades)
````

## Methods — the behaviour inside an object

A **method** is a function defined inside a class. It is called on an instance:

```python
amara.add_grade(88)
```

The difference from a plain function is that a method can read and change the
object's own attributes. The data and the code that maintains it now live in the
same place — which was the whole point.

Methods get docstrings too.

````python
class Student:
    """A single student, with a name and a list of grades."""

    def __init__(self, name):
        """Create a student with the given name and no grades yet."""
        self.name = name
        self.grades = []

    def add_grade(self, score):
        """Record one grade for this student."""
        self.grades.append(score)

    def average(self):
        """Return the mean grade, or 0.0 if there are no grades yet."""
        if len(self.grades) == 0:
            return 0.0
        return sum(self.grades) / len(self.grades)
````

````python
amara = Student("Amara")
amara.add_grade(91)
amara.add_grade(84)

liam = Student("Liam")
liam.add_grade(70)

print(f"{amara.name}: {amara.average():.1f}")
print(f"{liam.name}: {liam.average():.1f}")
````

## `self` — the number one confusion

Every method's first parameter is `self`. Two rules, and they explain everything:

1. **When you define** a method, you write `self` as the first parameter.
2. **When you call** it, you do *not* pass it. Python passes the object on the
   left of the dot.

```
amara.add_grade(91)      is really      Student.add_grade(amara, 91)
```

`self` is not a keyword. It is just the conventional name for "the instance this
method was called on." Always use `self`.

### Tracing `self` on two instances

The *same* method. Two different objects. Watch what `self` is each time.

````python
class Student:
    """A student that reports which object a method is running on."""

    def __init__(self, name):
        """Store the student's name and an empty grade list."""
        self.name = name
        self.grades = []

    def add_grade(self, score):
        """Add a grade and announce whose object self refers to."""
        print("self.name is", self.name, "- adding", score)
        self.grades.append(score)


amara = Student("Amara")
liam = Student("Liam")
````

````python
amara.add_grade(91)   # self is the Amara object
liam.add_grade(70)    # self is the Liam object
amara.add_grade(84)   # self is the Amara object again

print("Amara's grades:", amara.grades)
print("Liam's grades: ", liam.grades)
````

### Why `self` must be there

Without `self.`, a name inside a method is just a **local variable** — it
disappears when the method returns (scope, Week 4).

`self.grades` reaches *out* to the object and survives.

````python
class Cart:
    """A shopping cart demonstrating local vs. attribute assignment."""

    def __init__(self):
        """Start with an empty total."""
        self.total = 0.0

    def add_wrong(self, price):
        """Add to a LOCAL variable — the object is unchanged."""
        total = self.total + price   # local; thrown away

    def add_right(self, price):
        """Add to the object's own attribute — the change persists."""
        self.total = self.total + price
````

````python
cart = Cart()
cart.add_wrong(9.99)
print("after add_wrong:", cart.total)
cart.add_right(9.99)
print("after add_right:", cart.total)
````

## Objects vs dictionaries — when does a class earn its keep?

You can already store a student as a dictionary. So why a class?

**Rule of thumb:** a dictionary stores data. A class stores data **plus the rules
about that data**.

Let us build the *same* student both ways and compare the code that uses it.
This is the decision you must make — and justify — in your final project.

### Version A — dictionary + free functions

````python
def make_student(name):
    """Return a new student record as a dictionary."""
    return {"name": name, "grades": []}


def add_grade(student, score):
    """Append a grade to a student dictionary."""
    student["grades"].append(score)


def average(student):
    """Return the mean grade of a student dictionary."""
    if len(student["grades"]) == 0:
        return 0.0
    return sum(student["grades"]) / len(student["grades"])
````

````python
amara = make_student("Amara")
add_grade(amara, 91)
add_grade(amara, 84)

print(amara["name"], round(average(amara), 1))
print(amara)
````

### Version B — a class

````python
class Student:
    """A student with a name and a list of grades."""

    def __init__(self, name):
        """Create a student with no grades yet."""
        self.name = name
        self.grades = []

    def add_grade(self, score):
        """Record one grade for this student."""
        self.grades.append(score)

    def average(self):
        """Return the mean grade, or 0.0 if there are none."""
        if len(self.grades) == 0:
            return 0.0
        return sum(self.grades) / len(self.grades)
````

````python
amara = Student("Amara")
amara.add_grade(91)
amara.add_grade(84)

print(amara.name, round(amara.average(), 1))
````

### Comparing them honestly

| | Dictionary | Class |
| :--- | :--- | :--- |
| Access | `amara["name"]` — typo gives `KeyError` at runtime | `amara.name` — editor can autocomplete |
| Behaviour | `average(amara)` — function could be anywhere | `amara.average()` — attached to the data |
| Guarantees | any key may be missing | `__init__` guarantees every field exists |
| Reading it | `dict` — but a dict of *what*? | `Student` — the name says it |
| Cost | zero setup | a `class` block to write and maintain |

### So which one?

**Use a dictionary when:**
- the data is a plain lookup table (word → count, id → row),
- the keys are not known in advance or come from a file,
- there is no behaviour, only storage.

**Use a class when:**
- the same few fields always travel together,
- there are operations that only make sense on that data,
- you want one place to enforce the rules (a grade must be 0–100).

**In your project, be ready to say which and why.** "It felt right" is not an answer.

## Recap in one cell

Everything from Tuesday, in fifteen lines. This class is reused all session.

````python
class BankAccount:
    """A simple bank account with an owner and a balance."""

    def __init__(self, owner, balance=0.0):
        """Create an account for owner, optionally with a starting balance."""
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        """Add amount to the balance."""
        self.balance = self.balance + amount

    def withdraw(self, amount):
        """Remove amount if the funds are there; return True on success."""
        if amount > self.balance:
            return False
        self.balance = self.balance - amount
        return True
````

## Multiple instances, independent state

One class, many objects. Each instance has **its own copy** of every attribute set
in `__init__`.

This is the payoff. A single 15-line class describes ten thousand accounts, and
changing one never touches another.

The class is the blueprint; the instances are the buildings.

````python
accounts = [
    BankAccount("Priya", 120.0),
    BankAccount("Marcus", 40.0),
    BankAccount("Dana"),
]

accounts[0].deposit(30.0)
accounts[1].withdraw(15.0)

for account in accounts:
    print(account.owner, "->", account.balance)
````

Objects live happily inside lists and dictionaries. A list of `BankAccount`
objects is far clearer than three parallel lists of names, balances, and limits.

## The object lifecycle

1. **Creation** — `BankAccount("Dana")` allocates the object, then runs `__init__`.
2. **Use** — methods read and change its attributes; it survives between calls.
3. **Death** — when the last name referring to it disappears, Python reclaims it.

You almost never manage step 3 yourself. What matters practically: an object lives
exactly as long as something still refers to it — a variable, a list slot, a
dictionary value.

````python
by_owner = {}
temp = BankAccount("Dana", 10.0)
by_owner["Dana"] = temp     # the dictionary now also refers to it

temp = None                 # the variable lets go...
print(by_owner["Dana"].owner, by_owner["Dana"].balance)
# ...but the object is alive, because the dictionary still holds it.
````

## `__str__` — making objects readable

Print an object without help and you get something useless:

```
<__main__.BankAccount object at 0x10f3c9d50>
```

Define `__str__` and `print()` uses it instead. It must **return a string**, not
print one.

This costs three lines and makes every debugging session easier — worth it in
your project.

````python
class BankAccount:
    """A bank account that can describe itself as readable text."""

    def __init__(self, owner, balance=0.0):
        """Create an account for owner with an optional starting balance."""
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        """Add amount to the balance."""
        self.balance = self.balance + amount

    def __str__(self):
        """Return a human-readable description of this account."""
        return f"{self.owner}: ${self.balance:.2f}"
````

````python
dana = BankAccount("Dana", 10.0)
print(dana)              # print() calls __str__
dana.deposit(5.5)
print(dana)
print(str(dana).upper())  # str() calls it too
````

## A first look at inheritance

Sometimes a new class is *almost* an existing one, with a little extra.

**Inheritance** lets a class start from another class:

```python
class SavingsAccount(BankAccount):
    ...
```

- `BankAccount` is the **parent** (or base class).
- `SavingsAccount` is the **child** (or subclass).
- The child gets every method of the parent for free, and may add or replace some.

One parent, one child. That is all you need this term.

### The child inherits everything

````python
class SavingsAccount(BankAccount):
    """A bank account that also pays interest."""

    def add_interest(self, rate):
        """Increase the balance by rate (e.g. 0.02 for 2%)."""
        self.balance = self.balance + self.balance * rate


savings = SavingsAccount("Priya", 200.0)
savings.deposit(50.0)     # inherited from BankAccount
savings.add_interest(0.02)  # new in SavingsAccount
print(savings)            # __str__ inherited too
````

### Adding to the parent's `__init__`

If the child needs an extra attribute, write its own `__init__` and call the
parent's first with `super().__init__(...)`.

````python
class SavingsAccount(BankAccount):
    """A bank account with an interest rate of its own."""

    def __init__(self, owner, balance=0.0, rate=0.02):
        """Create a savings account, then store its interest rate."""
        super().__init__(owner, balance)   # let the parent set owner/balance
        self.rate = rate

    def add_interest(self):
        """Apply this account's interest rate to the balance."""
        self.balance = self.balance + self.balance * self.rate


savings = SavingsAccount("Priya", 200.0, 0.05)
savings.add_interest()
print(savings, "| rate:", savings.rate)
````

That single `super().__init__(...)` line is the only `super()` you need in COMP 1110.
It means: *"do the parent's setup, then continue with mine."*

### Overriding a method

A child may redefine a parent method. The child's version wins.

````python
class SavingsAccount(BankAccount):
    """A savings account that labels itself differently when printed."""

    def __str__(self):
        """Return a description marking this as a savings account."""
        return f"[SAVINGS] {self.owner}: ${self.balance:.2f}"


chequing = BankAccount("Marcus", 80.0)
savings = SavingsAccount("Marcus", 300.0)

for account in [chequing, savings]:
    print(account)    # each object uses its own class's __str__
````

## When *not* to use inheritance

Inheritance is the most over-used idea in OOP. At this level, use it only when
you can honestly say **"a child *is a kind of* parent."**

A `SavingsAccount` **is a** `BankAccount`. ✅

Do **not** inherit when:

- **It is "has-a", not "is-a".** A `Playlist` *has* `SongTrack`s — store a list of
  them; do not subclass.
- **You only want to reuse one function.** Write a plain function (Week 4).
- **The classes merely look similar.** `Student` and `Instructor` sharing a name
  field is not enough reason.

**For your project: one class is required. Zero subclasses is a perfectly good design.**

````python
# LIVE CODING — build this together.
#
# 1. A class Book with a docstring:
#      __init__(self, title, pages)
#      __str__(self)  -> "Dune (412 pages)"
#
# 2. A subclass AudioBook(Book):
#      __init__(self, title, pages, minutes)  -> use super()
#      __str__(self)  -> "Dune (412 pages, 320 min audio)"
#
# 3. Put one of each in a list and print them in a loop.
````

````python
class Book:
    """A printed book with a title and a page count."""

    def __init__(self, title, pages):
        """Create a book with the given title and number of pages."""
        self.title = title
        self.pages = pages

    def __str__(self):
        """Return a readable description of the book."""
        return f"{self.title} ({self.pages} pages)"


class AudioBook(Book):
    """A book that also has a recorded running time."""

    def __init__(self, title, pages, minutes):
        """Create an audiobook, reusing Book's setup for title and pages."""
        super().__init__(title, pages)
        self.minutes = minutes

    def __str__(self):
        """Return a description including the audio running time."""
        return f"{self.title} ({self.pages} pages, {self.minutes} min audio)"


library = [Book("Dune", 412), AudioBook("Dune", 412, 320)]
for item in library:
    print(item)
````

## Common errors (1 of 2)

**1. Forgetting `self` in the definition**

```python
def add_grade(score):     # missing self
```
```
TypeError: add_grade() takes 1 positional argument but 2 were given
```
Python passed the object automatically; your method had nowhere to put it.

**2. Forgetting `self.` inside the method**

Assigning `total = ...` instead of `self.total = ...` creates a local variable
that vanishes. No error at all — just a value that never changes.

````python
class Student:
    """Demonstrates the missing-self error."""

    def __init__(self, name):
        """Store the student's name."""
        self.name = name

    def greet():
        """This method is missing self and cannot be called on an instance."""
        return "hi"


s = Student("Amara")
s.greet()
````

## Common errors (2 of 2)

**3. Using an attribute `__init__` never created**

```
AttributeError: 'Student' object has no attribute 'grades'
```
Set every attribute in `__init__`, even to `[]`, `0`, or `None`.

**4. Calling the class instead of an instance (or the reverse)**

`Student.average()` without an instance gives a `TypeError` — there is no `self`.
And `amara.average` without parentheses prints `<bound method ...>` instead of a
number.

````python
class Student:
    """Demonstrates the missing-attribute error."""

    def __init__(self, name):
        """Store only the name — deliberately forgetting grades."""
        self.name = name


amara = Student("Amara")
print(amara.grades)
````
