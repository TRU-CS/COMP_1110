# Part 1 — Git and GitHub

## Why version control exists

You already know this problem:

```
project.py
project_v2.py
project_FINAL.py
project_FINAL_actually.py
project_FINAL_use_this_one_amina.py
```

That scheme fails in three ways: nobody can tell which file is current, nobody
can tell **what changed** between two of them, and when two people edit at once
one person's work quietly disappears.

**Git** fixes all three. It keeps one file — plus a complete, labelled history
of every version it has ever had.

## The four words you need today

| Word | What it means |
| :--- | :--- |
| **Repository** (*repo*) | A folder Git is watching, plus the full history of every change ever made to it. |
| **Commit** | One saved snapshot of the whole project, with a message saying what changed and who made it. |
| **Push** | Send your commits from your machine up to GitHub, so your team can see them. |
| **Pull** | Bring your teammates' commits down from GitHub onto your machine. |

**Git** is the program on your computer. **GitHub** is the website that stores a
copy your whole team can reach. They are not the same thing.

## What a commit actually is

A commit is **not** a backup of one file. It is a snapshot of the *whole
project* at one moment, carrying four pieces of information:

1. **What the files looked like** at that moment,
2. a **message** you wrote saying what changed,
3. **who** made it — your name and email, from your Git configuration,
4. **when** — the exact date and time.

Points 2, 3, and 4 are why the commit history is the evidence of who did what on
your team. Nothing else in your submission records that.

## Setting up — once per person, once per team

**Once per person**, so your commits carry your name:

```bash
git config --global user.name "Amina Chen"
git config --global user.email "you@mytru.ca"
```

**Once per team** — one member creates a **public** repository on GitHub, then
adds the other members as collaborators (*Settings → Collaborators*). Everyone
else copies it down:

```bash
git clone https://github.com/your-team/gradebook-analyzer.git
cd gradebook-analyzer
```

Use the email your commits should be traceable to. Commits made under a
mistyped email do not show up as yours.

## The everyday loop

Four commands, in this order, every time you sit down to work:

```bash
git pull                       # 1. get your teammates' latest work
                               # 2. ... edit your files ...
git status                     # 3. what have I changed?
git add scoreboard.py          # 4. choose what goes in this commit
git commit -m "Add average score calculation"
git push                       # 5. send it to GitHub
```

`git status` is the one to run when you are confused. It tells you what has
changed, what is staged, and usually suggests the command you want next.

### Predict the output (1)

You edited **two** files, then staged only one:

```bash
$ git add scoreboard.py
$ git status
```

You changed `scoreboard.py` **and** `README.md`. If you ran `git commit` right
now, which files would go into the commit?

**Answer — only `scoreboard.py`.**

`git status` reports the two files in two different sections:

```
Changes to be committed:
        modified:   scoreboard.py

Changes not staged for commit:
        modified:   README.md
```

`git add` is how you **choose** what belongs in the next commit. `README.md` is
changed on disk but was never added, so the commit ignores it — and your
teammate pulling your work would get the new code with no updated README.

This is the single most common way work goes missing: it was edited, but never
added.

## Commit messages

The message is written for the person reading the history later — a teammate, or
me marking your project.

**A good message:**

- says **what changed**, in the present tense: *"Add tie-break rule to the
  referee"*,
- is one line, roughly under 60 characters,
- describes **one** logical change.

**Commit as you go**, in pieces you can describe in one line. A single "final
project" commit at 11 PM on December 8 tells me nothing about who wrote what.

### Messages: before and after

| ❌ Not useful | ✅ Useful |
| :--- | :--- |
| `update` | `Add average score calculation` |
| `fixed it` | `Fix crash when the grades file is empty` |
| `asdf` | `Rename total to running_total for readability` |
| `changes from today` | `Add unit tests for is_passing()` |
| `final version` | `Handle missing file with try/except` |

Read the right-hand column top to bottom: it is a readable account of a project
being built. That is what a commit history is *for*.

### Predict the output (2)

Amina commits `Add score parser`, then Ben commits `Add average calculation`,
then Chen commits `Fix empty file crash`. Then anyone runs:

```bash
$ git log --oneline
```

Which of the three messages is printed **first**, and what does the short code
in front of each line tell you?

**Answer — `Fix empty file crash` is printed first.**

```
9c1f4ab Fix empty file crash
3ad77e0 Add average calculation
b12e905 Add score parser
```

`git log` prints **newest first**, so you read a history top-down as
*most recent → oldest*.

The short code is the commit's **hash** — a unique name for that snapshot. It
tells you *nothing* about ordering: `9c1f4ab` is not "after" `3ad77e0` because
of its characters. Order comes from the position in the list, not the hash.

Add `git log --oneline --author="Amina"` to see one person's commits only —
which is exactly what I do when I check contributions.

## Your commit history is the evidence

From the syllabus, and I mean it literally:

> **The commit history is the evidence of your contribution.** A member whose
> contribution does not appear in the history has no record of having
> contributed — whatever the team says afterwards.

So:

- **Everyone pushes their own work, under their own account.** One person
  pushing on everyone's behalf erases the other three from the record.
- **Do not paste a teammate's code into your own commit.** That credits you for
  their work and removes it from theirs.
- Commit in reasonable pieces as you go, not in one push at the end.

## Working as a team: pull before you push

Two people editing the same repository is the normal case, not an accident.

```bash
git pull      # always start here
# ... work, commit ...
git pull      # again, in case someone pushed while you worked
git push
```

If you `push` without pulling first, Git refuses:

```
! [rejected]  main -> main (fetch first)
```

That is Git protecting your teammate's work, not an error you caused. Run
`git pull`, then push again.

## Merge conflicts

When you and a teammate change **the same lines** of the same file, Git cannot
choose for you. It puts both versions in the file and asks you to decide:

```python
<<<<<<< HEAD
passing_grade = 60
=======
passing_grade = 50
>>>>>>> main
```

- Between `<<<<<<< HEAD` and `=======` is **your** version.
- Between `=======` and `>>>>>>>` is **their** version.

A conflict is not a disaster and not a telling-off. It is Git saying: *two
humans disagreed here, and only a human can settle it.*

### Repairing a conflict

You edit the file by hand until it says what you want, **markers and all
removed**:

```python
passing_grade = 60
```

Then finish the merge like any other change:

```bash
git add gradebook.py
git commit -m "Set passing grade to 60 after team discussion"
git push
```

The three marker lines are not code. If you leave any of them in, Python stops
at the first one with `SyntaxError: invalid syntax` — a real error that has
confused many students into thinking Git broke their program.

## What does *not* belong in the repository

Commit your **source code and your documentation**. Not everything else.

Create a file named `.gitignore` in the repository, listing what Git should
ignore:

```
__pycache__/
*.pyc
.DS_Store
.vscode/
venv/
```

Never commit passwords, API keys, or anything containing another student's
personal information. A commit is permanent and your repository is **public** —
deleting a file later does not remove it from the history.

## Common Git errors, and what they mean

| Message | What happened | What to do |
| :--- | :--- | :--- |
| `! [rejected] main -> main (fetch first)` | A teammate pushed after your last pull. | `git pull`, then push again. |
| `nothing to commit, working tree clean` | You have not changed anything since the last commit — or you saved in the wrong folder. | Check you edited the file inside the repo, and saved it. |
| `CONFLICT (content): Merge conflict in gradebook.py` | You and a teammate changed the same lines. | Edit the file, delete the `<<<<<<<` markers, `add`, `commit`. |
| `fatal: not a git repository` | You are in the wrong folder. | `cd` into the cloned repository first. |

Read the last line of any Git message first. Like Python, Git puts the summary
at the bottom.

```python
def count_down(start, step):
    total = 0
    for number in range(start, 0, -step):
        total = total + number
        print(number, total)
    return total


result = count_down(10, 3)

```

Exactly what is printed, and how many lines?

**Answer — 5 lines:**

```
10 10
7 17
4 21
1 22
result: 22
```

`range(10, 0, -3)` counts **down** from 10 while staying above 0: 10, 7, 4, 1.
It stops before 0 — `stop` is never included, and −2 would be past it anyway.

`total` accumulates as it goes: 10, then 17, then 21, then 22. The `return`
hands 22 back to `result`, which the last `print` shows.
