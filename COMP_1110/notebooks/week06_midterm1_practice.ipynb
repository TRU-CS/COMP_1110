## Part A practice — values, types and operators

Part A questions are one line each. Work them out in your head, then check.

Quick reminders:

- `/` **always** gives a `float`; `//` is floor division; `%` is the remainder;
  `**` is a power and groups **right to left**.
- Precedence: `**`, then `* / // %`, then `+ -`, then comparisons, then `not`,
  `and`, `or`.
- `print(a, b)` puts **one space** between items; `"a" + "b"` puts **nothing**.

### A1 — floor division and remainder

```python
print(15 // 4, 15 % 4)
```

What does this print?

- **A.** `3.75 3`
- **B.** `3 0.75`
- **C.** `4 3`
- **D.** `3 3`

<details>
<summary><b>Answer</b></summary>

**D.** `15 // 4` is `3` (4 fits into 15 three whole times) and `15 % 4` is `3`
(15 = 3×4 + **3**). Both operands are `int`, so both results are `int`.
**A** confuses `//` with `/`; **B** treats `%` as "the decimal part"; **C** rounds
instead of flooring.

</details>

## Part A practice — f-strings

An f-string is a string with an `f` in front. Anything inside `{ }` is
**evaluated as Python** and its value is put into the text.

```python
item = "pens"
print(f"{2 + 1} {item}")      # 3 pens
```

### B1 — an elif chain

</details>

### B9 — parameters are local


- **B.** `60`
- **C.** `10`

</details>

```python
def temperature_log():
    count = 0
    highest = None
    lowest = None
```
