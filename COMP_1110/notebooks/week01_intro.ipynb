## What is a program?

> A **program** is a sequence of instructions that specifies how to perform a computation.

Two ideas hiding in that sentence:

- **Sequence** — the instructions run in order, top to bottom, one at a time.
- **Specifies** — the computer does *exactly* what you wrote, not what you meant.

Programming is really two skills:

1. Knowing the **language** (Python's vocabulary and grammar) — a few weeks of work.
2. Knowing how to **decompose a problem** into steps a machine can follow — the rest of
   your career.

## A program is just steps, in order

"Print a greeting, then work out how much a 3-item lunch costs."

````python
# Three instructions. Python runs them top to bottom, one at a time.
print("Lunch receipt")
print("Sandwich, coffee, cookie")
print(6.50 + 2.25 + 1.75)
````

Notice: nothing is *displayed* unless you ask with `print()`. The computer is not being
helpful — it is being literal.

## The machine underneath

| Part | What it does | Rough analogy |
| :--- | :--- | :--- |
| **CPU** | Executes instructions — billions per second, but only very simple ones | the worker |
| **Main memory (RAM)** | Holds the program and its data *while it runs*; **forgotten** at power-off | the desk |
| **Secondary memory** | Disk / SSD — survives power-off | the filing cabinet |
| **Input devices** | Keyboard, mouse, camera, microphone | the ears and eyes |
| **Output devices** | Screen, speakers, printer | the mouth |
| **Network** | Another kind of very slow input/output | the telephone |

The CPU is **fast and stupid**. Your job is to give it instructions so precise that being
stupid does not matter.

## Interpreter vs compiler

The CPU only understands **machine language** — numbers. You write **Python**, a
*high-level language*. Something has to bridge the gap. There are two strategies:

- An **interpreter** reads your source code a bit at a time, translates it, and
  **runs it immediately**. Python works this way.
- A **compiler** reads the *whole* source file, translates it into a machine-language
  **executable file**, and that file is run later. C and Java work this way.

Interpreter → you get an answer instantly, and errors show up as you reach them.
Compiler → slower to start, faster to run, and many errors are caught before it ever runs.

## Two ways to hand code to the Python interpreter

**1. Interactively.** Type `python3` in a terminal, get a `>>>` prompt, type one line,
get an answer instantly:

```text

```

Great for experiments. **Nothing is saved.**

**2. As a script.** Put the lines in a file called `lunch.py`, then run `python3 lunch.py`.
Saved, repeatable, submittable — this is what worksheets want.

## The building blocks of every program

Every program you will ever write — in any language — is built from six things:

| Block | What it means | We cover it |
| :--- | :--- | :--- |
| **Input** | Get data from keyboard, file, or network | Week 2 |
| **Output** | Show results on screen or write to a file | **today** |
| **Sequential execution** | Run statements in the order they appear | **today** |
| **Conditional execution** | Check a condition, skip or run some code | Week 3 |
| **Repetition** | Do something many times, usually with variation | Week 5 |
| **Reuse** | Write a chunk once, name it, call it many times | Week 4 |

That is the whole list. The rest of the term is practice at combining them.

Today we use exactly two of them: **output** and **sequential execution**. That is already
enough to write a real, useful program.

## Your first statement: `print()`

`print()` is a **function**: a name, then parentheses. What goes inside is the **argument**.

- Quotes make a **string** (text). `'` or `"` — but they must match.
- The parentheses are **not optional**: `print "hi"` is Python 2 and errors in Python 3.

````python
print("Hello, COMP 1110!")
print('Single quotes work exactly the same way.')
print()  # no argument: prints a blank line
print("Room OM 1241, Tuesdays and Thursdays.")
````

## `print()` with several arguments

Separate arguments with commas. Python prints them in order with **one space between**.

````python
print("Sandwich", 6.50)
print("Total items:", 3)
print("High", "temperature", "today:", 27, "degrees")
````

A comma inserts a space and accepts **any** type. A `+` joins strings and does **not** add a
space — and it refuses to mix text with numbers. Comma is the safer habit for now.

## Predict the output — 2

```python
print("Grade:", 88)
print("Grade:" + " " + "88")
print("Grade")
print("88")
```

How many **lines** appear, and what is on each?

````python
print("Grade:", 88)
print("Grade:" + " " + "88")
print("Grade")
print("88")
````

**Four lines** — one per `print()` call. The first two look identical, but the first joins a
string to an *integer* using a comma; the second joins three *strings* with `+`. Each
`print()` ends its line automatically.

## Python as a calculator

The interpreter evaluates arithmetic **expressions** before printing them.

| Operator | Meaning | Example | Result |
| :--- | :--- | :--- | :--- |
| `+` `-` `*` | add, subtract, multiply | `4 * 3` | `12` |
| `/` | divide — **always gives a decimal** | `7 / 2` | `3.5` |
| `**` | power | `2 ** 10` | `1024` |

Normal precedence applies: `**` first, then `*` and `/`, then `+` and `-`. Parentheses
override it. When in doubt, add parentheses — they cost nothing and save arguments.

````python
print(6.50 + 2.25 + 1.75)      # a lunch total
print(2 ** 10)                 # 2 to the power of 10
print(85 + 91 + 78)            # three test scores
print((85 + 91 + 78) / 3)      # ...and their average
````

## Predict the output — 3

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(10 / 4)
```

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(10 / 4)
````

**Why:** `*` binds tighter than `+`, so line 1 is `2 + 12`. Parentheses force the addition
first in line 2. And `/` in Python 3 **always** produces a decimal — `2.5`, never `2`.

## What can go wrong: three kinds of error

| Error | When it happens | Python's reaction |
| :--- | :--- | :--- |
| **Syntax error** | Your code breaks Python's grammar | Refuses to run **anything**; reports a line number |
| **Logic error** | Grammar fine, but the *order* or *arithmetic* of your steps is wrong | Runs happily, gives a **wrong answer** |
| **Semantic error** | Grammar fine, runs fine, but it does something other than what you **meant** | Silence — you must notice |

**Only the first kind gets you an error message.** The other two are found by reading,
tracing, and testing — which is why every worksheet has a trace-and-explain question.

## A syntax error

A missing closing parenthesis. Python cannot even begin.

````python
print("Missing a closing parenthesis"
````

`SyntaxError` — read the message **bottom-up**: the last line names the error, the line
above points at where Python gave up. It often points at the line *after* the real mistake,
because that is where the problem became visible.

## A logic error and a semantic error

Both of these run without a single complaint.

````python
# LOGIC error: the average of three scores, divided too early.
print(85 + 91 + 78 / 3)     # prints 202.0, not 84.66...

# SEMANTIC error: this runs, but it labels the wrong thing.
print("Average:", 85 + 91 + 78)
````

Line 1 is a **logic** error: precedence divided only `78`. Line 2 is a **semantic** error:
the arithmetic is a correct *sum*, but the program claims it is an *average*. Python cannot
help you with either.

## Common errors in Week 1

| What you type | What Python says | What it means |
| :--- | :--- | :--- |
| `print "hi"` | `SyntaxError: Missing parentheses in call to 'print'` | Python 2 style. Add the parentheses. |
| `print(hello)` | `NameError: name 'hello' is not defined` | No quotes — Python looked for a *thing* named `hello`. |
| `print("hi')` | `SyntaxError: unterminated string literal` | Opening and closing quotes must **match**. |
| `Print("hi")` | `NameError: name 'Print' is not defined` | Python is **case-sensitive**. It is `print`. |

````python
print(hello)
````

`NameError` is Python saying *"I have never heard of that word."* Nine times out of ten in
Week 1, it means you forgot the quotation marks.
