## PEP 8: a shared convention

**PEP** = *Python Enhancement Proposal*. PEP 8 is the style guide almost
every Python project follows. Its value is not that its rules are the best
possible rules — it is that **everyone already uses them**.

| Area | Rule |
| :--- | :--- |
| Indentation | 4 spaces per level, never tabs |
| Naming | `snake_case` functions/variables · `CapWords` classes · `UPPER_CASE` constants |
| Spacing | Spaces around `=`, `+`, `<` · a space after every comma |
| Line length | Under ~79 characters |
| Blank lines | 2 between top-level definitions, 1 to group logic inside |
| Imports | Top of the file, one per line |
| Comments | Explain *why*, never restate *what* |

## Naming: the highest-value rule

A name is the cheapest documentation you will ever write, and the only one
that gets read every single time.

**A name should say what the thing IS.**

- `n` → a count? a name? a number?
- `data` → says nothing; all variables hold data
- `temp` → temperature, or temporary?
- `student_count` → type, purpose, and plural-ness, all for free

Read booleans as a question (`is_registered`), functions as the value they
return (`average_grade`) or the action they take (`print_report_card`).

If you cannot name it, you do not yet understand it.

### The four conventions

````python
MAX_CLASS_SIZE = 40          # constant: UPPER_CASE with underscores


class StudentRecord:         # class: CapWords, no underscores
    """One student's name and grade."""

    def __init__(self, full_name, grade):
        self.full_name = full_name    # attribute: snake_case
        self.grade = grade


def is_passing(record):      # function: snake_case, reads as a question
    """Return True if the record's grade is 50 or more."""
    return record.grade >= 50


top_student = StudentRecord("Priya Sharma", 91)   # variable: snake_case
print(top_student.full_name, is_passing(top_student), MAX_CLASS_SIZE)
````

You did not have to read a single function body to know what these do.
That is the whole point.

## Indentation and whitespace

**4 spaces** per level — not 2, not 8, not a tab. Never mix tabs and spaces
in one file; Python 3 refuses to run it. Set your editor to insert spaces
when you press Tab (VS Code does by default).

Whitespace *inside* a line is not syntax, but it is how the eye finds the
operators:

````python
# GOOD: spaces around operators, after commas, none inside brackets
subtotal = 19.99 + 4.50
tax = round(subtotal * 0.12, 2)
cart = ["milk", "bread", "apples"]
print(cart[0], subtotal, tax, sep=" | ")   # no spaces around = for keywords
````

**⚠️ COUNTER-EXAMPLES.** All legal Python, all wrong:
`subtotal=19.99+4.50` · `round( subtotal*0.12,2)` · `[ "milk","bread" ]` ·
`print (cart[0])`

## Line length

PEP 8 says **79 characters**; many teams allow 88 (`black`'s default).
Long lines scroll sideways in a diff, stop two files fitting side by side,
and lose the eye on the way back. A too-long line usually means the line is
doing too much.

**Break inside brackets that are already there** — `(`, `[`, `{`.

````python
students = ["Amrit", "Bao", "Chen", "Dania", "Elif", "Farid"]

message = "Enrolled: " + ", ".join(
    students[:4]
)
print(message)

count = len(students)
if (count > 3
        and count < 10):
    print("Good class size:", count)
````

Avoid ending a line with a backslash: one invisible space after it is a
silent syntax error. Brackets never lie.

## Layout: blank lines and imports

**Blank lines are paragraph breaks.** Two between top-level definitions,
one between methods, one inside a function to separate logical steps —
and *zero* just because. Double-spacing everything destroys the signal.

**Imports go at the top of the file**, after the module docstring, one
module per line, standard library first. A reader should see every
dependency in the first ten lines without scrolling.

````python
"""Report on a small set of exam marks."""

import math


def summarise(readings):
    """Print the count, average, and hottest of a list of temperatures."""
    count = len(readings)
    total = sum(readings)

    average = total / count
    hottest = max(readings)

    print("Readings:", count, "| avg:", round(average, 1), "| max:", hottest)


summarise([18.5, 22.0, 25.5, 19.0])
print(math.floor(9.7))
````

Three visual blocks inside the function: **gather**, **compute**, **report**.
You can see its shape before reading a line of it.

**⚠️ COUNTER-EXAMPLES:** `import math, random` on one line; an `import`
buried inside a loop. Both run, both hide information.

## Comments that earn their place

A comment costs something: it can go stale, and then it actively **lies**.

- A good comment explains **WHY**.
- A bad comment restates **WHAT** the code already says.

```python
count = count + 1   # add one to count                    <- worthless
count = count + 1   # the header line is not a data row   <- earns its place
```

**⚠️ INTENTIONAL COUNTER-EXAMPLE — this cell breaks PEP 8 on purpose. Do not imitate it.**

````python
# set the total to zero
total = 0
# make a list of prices
prices = [4.99, 12.50, 3.25]
# loop over the prices
for price in prices:
    # add the price to the total
    total = total + price
# print the total
print(total)
````

Delete every comment and the code gets **easier** to read: half the text,
and nothing left that can quietly become out of date.

### When a comment should have been a better name

````python
# BEFORE: the comment is propping up a bad name
# d is the number of days until the assignment is due
d = 7
if d < 3:
    print("Start now.")

# AFTER: the name carries the meaning; the comment is not needed
days_until_due = 7
if days_until_due < 3:
    print("Start now.")
````

**If a comment explains what a variable means, rename the variable.** The
name travels with the value everywhere it is used; the comment stays on
line 4 and rots.

### When a comment should have been a function

````python
readings = [18.5, 22.0, 25.5]

# BEFORE: a comment announcing a block is a function trying to be born
# --- convert celsius to fahrenheit and print ---
for celsius in readings:
    print(round(celsius * 9 / 5 + 32, 1))


# AFTER: the block has a name, so it can be tested and reused
def print_in_fahrenheit(celsius_readings):
    """Print each Celsius reading converted to Fahrenheit."""
    for celsius in celsius_readings:
        print(round(celsius * 9 / 5 + 32, 1))


print_in_fahrenheit(readings)
````

A comment of the form `# now do X` is almost always a function called
`do_x()` waiting to be extracted. Hunt for them in your project this week.

### Fix 2 — indentation and spacing

````python
def average_of_passing(marks):
    total = 0
    passing_count = 0
    # loop
    for mark in marks:
        if mark >= 50:
            total = total + mark
            passing_count = passing_count + 1
    if passing_count == 0:
        return 0
    return total / passing_count
class_marks = [45, 72, 88, 30, 91]
print(round(average_of_passing(class_marks), 2))
````

4-space indent makes the nesting depth obvious at a glance; spaces around
operators show where each expression starts and stops.

### Fix 3 — kill the useless comment, add blank lines

````python
def average_of_passing(marks):
    total = 0
    passing_count = 0

    for mark in marks:
        if mark >= 50:
            total = total + mark
            passing_count = passing_count + 1

    if passing_count == 0:
        return 0
    return total / passing_count


class_marks = [45, 72, 88, 30, 91]
print(round(average_of_passing(class_marks), 2))
````

`# loop` is gone — `for mark in marks:` already said that. Three paragraphs
now: **set up**, **accumulate**, **decide what to return**. Two blank lines
before the top-level code, as PEP 8 asks.

### Fix 4 — the missing docstring

````python
def average_of_passing(marks):
    """Return the average of the marks that are 50 or above.

    Returns 0.0 when no mark reaches 50, so the caller never
    divides by zero.
    """
    total = 0
    passing_count = 0

    for mark in marks:
        if mark >= 50:
            total = total + mark
            passing_count = passing_count + 1

    if passing_count == 0:
        return 0.0
    return total / passing_count


print(round(average_of_passing([45, 72, 88, 30, 91]), 2))
````

The docstring documents the **edge case** — the one thing a reader genuinely
could not work out in two seconds. That is a comment earning its place, and
it is required on every function in your project.

### Version 4 — the finished program

````python
PASSING_MARK = 50


def average_of_passing(marks):
    """Return the average of marks >= PASSING_MARK, or 0.0 if there are none."""
    passing = []
    for mark in marks:
        if mark >= PASSING_MARK:
            passing.append(mark)

    if len(passing) == 0:
        return 0.0
    return sum(passing) / len(passing)


assert average_of_passing([]) == 0.0
assert average_of_passing([50, 60]) == 55.0

print(round(average_of_passing([45, 72, 88, 30, 91]), 2))
````

Notice what became possible only *after* cleaning up: the magic number `50`
got a name, and the function got clear enough to write `assert` tests for.
**Readable code is testable code.**

````python
def clean_name(raw_name):
    """Return the name stripped of spaces and in title case."""
    return raw_name.strip().title()


names = ["  ada lovelace ", "GRACE HOPPER"]
for name in names:
    print(clean_name(name) + "!")
````

**Answer:** `Ada Lovelace!` then `Grace Hopper!`

`.strip()` removes leading and trailing spaces; `.title()` upper-cases the
first letter of each word and **lower-cases the rest** — which is why the
shouted `GRACE HOPPER` comes back as `Grace Hopper`. The docstring told you
both steps before you read the body.

## Tooling: stop doing this by hand

Nobody memorises all of PEP 8. We let the computer check it.

| Tool | What it does |
| :--- | :--- |
| **Linter** (`flake8`, `pycodestyle`) | **Reads** your file and *reports* problems. Changes nothing. |
| **Auto-formatter** (`black`) | **Rewrites** your file into a standard layout. |

```bash
pip install flake8
flake8 my_project.py
```
```
my_project.py:3:1:  E302 expected 2 blank lines, found 1
my_project.py:7:12: E225 missing whitespace around operator
my_project.py:9:80: E501 line too long (94 > 79 characters)
my_project.py:2:1:  F401 'random' imported but unused
```

`E###` = style. `F###` = a real problem. **Fix the `F` ones first.**

### An auto-formatter: `black`

```bash
pip install black
black my_project.py          # rewrites the file in place
black --check my_project.py  # only reports whether it would change anything
```

`black` has almost no options. That is the feature, not a limitation.

```python
x = {  'a':37,'b':42,
'c':927}                          # before
x = {"a": 37, "b": 42, "c": 927}  # after
```

Run it before every commit, and before you submit.

**Formatting is not worth arguing about.** On a team of four, "where do the
spaces go" has zero educational value and infinite capacity for argument.
Agree on `black` on day one, everyone runs it before pushing, never discuss
spacing again — and spend the hours you saved on **correctness and tests**.

*(PEP 8 has a section called "A Foolish Consistency is the Hobgoblin of
Little Minds": break a rule when following it would make code less readable.
Knowing the rule and choosing to break it is style. Not knowing it is not.)*

The cell below works and is unreadable. **We fix it in class**, one flaw at

````python
# TODO (in class): refactor into readable, PEP 8 code. Keep behaviour
# identical. It reports the percentage of text messages longer than
# 20 characters.

def g(m):
 n=0
 for x in m:
  if len(x)>20:n=n+1
 return n*100/len(m)
msgs=["hey","are we still meeting at four today?","ok","running late!!"]
print(round(g(msgs),1))
````

````python
# INSTRUCTOR SOLUTION (hidden from the deck)
LONG_MESSAGE_LENGTH = 20


def percent_long_messages(messages):
    """Return the percentage of messages longer than LONG_MESSAGE_LENGTH.

    Returns 0.0 for an empty list rather than dividing by zero.
    """
    if len(messages) == 0:
        return 0.0

    long_count = 0
    for message in messages:
        if len(message) > LONG_MESSAGE_LENGTH:
            long_count = long_count + 1

    return long_count * 100 / len(messages)


assert percent_long_messages([]) == 0.0
assert percent_long_messages(["short"]) == 0.0

texts = ["hey", "are we still meeting at four today?", "ok", "running late!!"]
print(round(percent_long_messages(texts), 1))
````

## Common errors this week

**1.** `TabError: inconsistent use of tabs and spaces in indentation`
You pasted code that used tabs into a file that uses spaces. In VS Code:
command palette → *Convert Indentation to Spaces*.

**2.** `IndentationError: unexpected indent`
A line is indented further than the block it belongs to. Look at the line
number **and the line above it**.

**3.** `flake8: command not found` — you installed into a different Python
than the one you are running. Use `python -m flake8 myfile.py`.

**4. `black` reformatted the file but `flake8` still complains.**
`black` fixes *layout* only. It will not fix a long string (`E501`), an
unused import (`F401`), or a bad variable name. Those are yours.

**5. The error you will actually meet on Dec 8:** a `SyntaxError` at
11:47 PM because someone pushed a half-finished edit. Freeze the code the
night **before**, run the whole program from a clean folder, run your tests,
submit, then stop touching it.
