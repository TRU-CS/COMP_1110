## Why not just use a list?

Last week we counted things with lists. Suppose we want to count how many times
each word appears in a text message.

With a list you must store the word **and** its count, then **search** for the
word every time:

- Where is `"the"`? You do not know — you must scan.
- Position is meaningless: item `3` tells you nothing about *what* it is.

A **dictionary** lets you look things up **by name** instead of by position.

````python
# The list version of "count words" is painful.
words = ["dog", "cat", "dog"]
counts = [0, 0]          # counts[0] is for "dog"... you have to REMEMBER that
labels = ["dog", "cat"]  # ...so you need a second, parallel list

for word in words:
    position = labels.index(word)   # a SEARCH on every single word
    counts[position] = counts[position] + 1

print(labels)
print(counts)
````

Two lists that must stay lined up, plus a search on every step. This is exactly
the problem a dictionary removes.

## A dictionary maps keys to values

A **list** maps *integer positions* → values.
A **dictionary** maps *keys you choose* → values.

```python
grades = {"Amina": 88, "Ben": 72, "Chen": 95}
#          ^key      ^value
```

- Written with **curly braces** `{}` and `key: value` pairs.
- You look things up with **square brackets and a key**: `grades["Ben"]`.
- Keys are usually strings or numbers. **Keys must be unique.**
- The value can be anything: a number, a string, even a list.

````python
grades = {"Amina": 88, "Ben": 72, "Chen": 95}

print(grades)
print(grades["Ben"])       # look up by KEY, not by position
print(len(grades))         # how many key-value pairs

empty = {}                 # an empty dictionary
print(empty, len(empty))
````

## Adding and updating entries

The same syntax does both:

```python
grades["Dara"] = 60     # key not there  -> ADDS a new pair
grades["Ben"]  = 75     # key already there -> REPLACES the old value
```

A dictionary is **mutable**, like a list. There is no `append()` — assigning to
a new key *is* the append.

Deleting: `del grades["Dara"]`.

````python
grades = {"Amina": 88, "Ben": 72}

grades["Dara"] = 60        # add
grades["Ben"] = 75         # update (72 is gone forever)
print(grades)

del grades["Amina"]        # remove a pair
print(grades)
print(len(grades))
````

## ⚠️ Common error: `KeyError`

Looking up a key that is **not in the dictionary** is an error — it does not
return `0`, and it does not return `None`. It **crashes the program**.

This is the single most common dictionary bug, and it is the reason `in` and
`get()` exist.

````python
grades = {"Amina": 88, "Ben": 72}
print(grades["Erik"])
````

```
KeyError: 'Erik'
```

**Read it as:** "you asked me for the key `'Erik'` and I do not have it."

Note what it is *not*: not an `IndexError` (that is lists), and not a
`NameError`. If you see `KeyError`, the quoted thing is a **key you assumed was
there**.

Three fixes, coming up next: check with `in`, use `get()`, or put the key there
first.

## The `in` operator on dictionaries

`in` asks: **is this a key?**

```python
if "Ben" in grades:
    print(grades["Ben"])
```

⚠️ Careful — on a dictionary, `in` searches the **keys only**, never the values.
`72 in grades` is `False` even when Ben's grade is 72.

Compare with a list, where `in` searches the *elements*.

````python
grades = {"Amina": 88, "Ben": 72}

print("Ben" in grades)        # True  -> it is a key
print("Erik" in grades)       # False -> not a key
print(72 in grades)           # False -> 72 is a VALUE, not a key
print("Erik" not in grades)   # True

if "Erik" in grades:
    print(grades["Erik"])
else:
    print("No grade recorded for Erik")
````

## `get()` — a lookup with a default

`get()` does the `if ... in ...` check for you, in one expression:

```python
grades.get("Erik", 0)
```

means: *"give me the value for `'Erik'`; if that key is missing, give me `0`
instead."* It **never raises `KeyError`**.

The default is whatever you pass — `0`, `""`, `"unknown"`. If you leave it out
you get `None`, which is rarely what you want.

````python
grades = {"Amina": 88, "Ben": 72}

print(grades.get("Ben", 0))       # 72  -> key found, default ignored
print(grades.get("Erik", 0))      # 0   -> key missing, default used
print(grades.get("Erik"))         # None -> no default given

# get() does NOT add anything to the dictionary:
print(grades)
````

## 🌟 The counting pattern (histogram)

**The most important idiom in Chapter 9.** Count how many times each item
appears in a sequence.

The dictionary is used as a **set of counters**: one counter per distinct key,
created on demand.

We will build it in **three stages** — same result, progressively shorter.

**Stage 1: the honest, long version** — check `in` first.

````python
words = ["the", "dog", "the", "cat", "the", "dog"]
counts = {}

for word in words:
    if word not in counts:
        counts[word] = 1        # first sighting: start the counter at 1
    else:
        counts[word] = counts[word] + 1   # seen before: bump it

print(counts)
````

Trace `"the"`: not in `counts` → set to `1`. Later `"the"` again → `1 + 1 = 2`.
Third time → `3`.

Write this version until your fingers know it. The next two are just shorthand.

## Stage 2: replace the `if` with `get()`

The `if` exists only to handle "the counter does not exist yet". That is
exactly what a default is for.

```python
counts[word] = counts.get(word, 0) + 1
```

*"Take the current count — or `0` if there isn't one — add 1, store it back."*

One line replaces four. **Learn this line.**

````python
words = ["the", "dog", "the", "cat", "the", "dog"]
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
print(counts["the"])
````

## Stage 3: the final, reusable form

Wrap it in a function so you can point it at anything — words, letters, grades,
song plays.

Note the **docstring** (Week 8) and the PEP 8 naming. This is the version to put
in your final project.

````python
def count_items(items):
    """Return a dictionary mapping each item in items to how often it appears."""
    counts = {}
    for item in items:
        counts[item] = counts.get(item, 0) + 1
    return counts


print(count_items(["the", "dog", "the"]))
print(count_items("mississippi"))      # a string is a sequence of characters too
````

## Looping over a dictionary

A `for` loop over a dictionary walks the **keys**:

```python
for name in grades:      # name takes each KEY in turn
    print(name, grades[name])
```

Three explicit methods say what you mean:

| Method | Gives you |
| :--- | :--- |
| `.keys()` | the keys |
| `.values()` | the values |
| `.items()` | key **and** value together |

````python
grades = {"Amina": 88, "Ben": 72, "Chen": 95}

for name in grades:                 # keys, by default
    print(name, "scored", grades[name])

print(list(grades.keys()))
print(list(grades.values()))
````

````python
grades = {"Amina": 88, "Ben": 72, "Chen": 95}

# .items() gives you both at once -- two loop variables
for name, score in grades.items():
    print(f"{name:<8} {score:>3}")

# .values() makes whole-dictionary arithmetic easy
total = 0
for score in grades.values():
    total = total + score
print("Class average:", total / len(grades))
````

Each pair from `.items()` is a **tuple** — formally Chapter 10, next week. For
today, read `for name, score in grades.items():` as "unpack the pair into two
variables".

## Dictionaries and files: counting words in real text

The classic Chapter 9 program:

1. **Open** a text file.
2. For each line, `split()` it into a list of words (Week 9).
3. Feed every word into the counting pattern.
4. Find the most common word.

First — so this notebook needs no outside data — we **write the file
ourselves**, then read it back.

````python
# Create the data file this notebook will read (Week 8: writing files).
text = """the quick brown fox jumps over the lazy dog
the dog barks and the fox runs away
a lazy dog sleeps in the sun
"""

handle = open("week10_text.txt", "w")
handle.write(text)
handle.close()

print("wrote week10_text.txt")
````

````python
counts = {}

handle = open("week10_text.txt")
for line in handle:
    words = line.split()            # a list of words on this line
    for word in words:
        counts[word] = counts.get(word, 0) + 1
handle.close()

print(len(counts), "distinct words")
print(counts["the"], "occurrences of 'the'")
print(counts)
````

A **nested loop**: the outer one walks lines, the inner one walks the words of
one line. The counting pattern is untouched — it does not care where the words
came from.

## Finding the most common word

We cannot sort a dictionary yet — `sorted()` is next week. But we already know
the **max pattern** from Week 5: keep a "best so far" and beat it.

````python
best_word = None
best_count = 0

for word, count in counts.items():
    if count > best_count:
        best_word = word
        best_count = count

print("Most common word:", best_word, "->", best_count)
````

## Advanced text parsing 🧹

Raw text is dirty. `"Dog"`, `"dog"` and `"dog."` are three different keys, which
means three wrong counters.

Standard clean-up, using only string methods from Week 7:

1. `line.lower()` — case no longer splits counts.
2. Remove punctuation, character by character.
3. *Then* `split()`.

````python
def clean(line):
    """Return line lowercased with punctuation removed."""
    punctuation = ".,;:!?\"'()-"
    line = line.lower()
    for mark in punctuation:
        line = line.replace(mark, " ")
    return line


print(clean("The dog barked; the DOG ran. Fast!"))
print(clean("The dog barked; the DOG ran. Fast!").split())
````

````python
messy = "Hello there! Hello, hello -- how are you? Are you well?"

counts = {}
for word in clean(messy).split():
    counts[word] = counts.get(word, 0) + 1

print(counts)
print("'hello' appears", counts["hello"], "times")
````

## ⚠️ Dictionary ordering — do not rely on it

Your textbook says dictionary order is **unpredictable**, and that is the rule
to program by.

Modern Python (3.7+) does happen to keep **insertion order**, so printing looks
stable on your machine — but:

- The order is **insertion order**, *never* alphabetical and *never* by value.
- Your program's meaning must not depend on it. "The first key" is not a
  meaningful idea.
- Two dictionaries with the same pairs in different order are **equal**:
  `{"a": 1, "b": 2} == {"b": 2, "a": 1}` is `True`.

Want a real order? Sort it — `sorted()`, **next week**.

````python
first = {"a": 1, "b": 2}
second = {"b": 2, "a": 1}

print(first == second)       # True -- order is not part of the value
print(first)
print(second)

# Alphabetical output WITHOUT sorted(): loop over a list you control.
for key in ["a", "b"]:
    print(key, first[key])
````

## 🎯 Dictionary or list? (this is examinable — CLO 3)

| Ask yourself | Use a **list** | Use a **dictionary** |
| :--- | :--- | :--- |
| How do I find an item? | by **position** (`items[0]`) | by **name/key** (`ages["Ben"]`) |
| Does order matter? | yes — first, last, 3rd | no |
| Are there duplicates? | yes, allowed | keys must be **unique** |
| What is the data? | a **sequence** of similar things | a **lookup table** / labelled facts |
| Typical job | queue of scores, lines of a file | counters, phone book, settings |

**One-sentence test:** *if you would naturally say "the value **for** X", it is a
dictionary; if you would say "the **next** one", it is a list.*

**Practise the exam question.** Which structure, and why?

1. The daily high temperature for each city. → **dict** (`temps["Kamloops"]`)
2. The temperatures for one city, in order, over 7 days. → **list** (order is the meaning)
3. How many times each student submitted late. → **dict** (a counter per key)
4. The lines of a file, to be printed back in order. → **list**
5. A password for each username. → **dict** (unique key → one value)

Expect exactly this on Midterm 2 — and in your project Q&A ("*why a dictionary
here rather than a list of pairs?*").

## 💻 Live coding 1 — grade book

````python
# TASK: build a grade book with a dictionary.
#  1. Start from the list of (name, score) data below.
#  2. Build a dictionary mapping name -> score.
#  3. Print the average score.
#  4. Ask for one name that IS present and one that is NOT,
#     printing "no record" instead of crashing.

names = ["Amina", "Ben", "Chen", "Dara"]
scores = [88, 72, 95, 60]

# your code here
````

````python
# Solution
names = ["Amina", "Ben", "Chen", "Dara"]
scores = [88, 72, 95, 60]

grades = {}
for position in range(len(names)):
    grades[names[position]] = scores[position]

total = 0
for score in grades.values():
    total = total + score
print("Average:", total / len(grades))

for who in ["Chen", "Erik"]:
    if who in grades:
        print(who, grades[who])
    else:
        print(who, "-> no record")
````

````python
# Solution
def count_letters(text):
    """Return a dictionary of letter -> count, ignoring case and spaces."""
    counts = {}
    for letter in text.lower():
        if letter != " ":
            counts[letter] = counts.get(letter, 0) + 1
    return counts


print(count_letters("Hello there"))
````

## Common errors 🐞

**1. `KeyError` — asking for a key that is not there.** (See the slide earlier.)
Fix with `in` or `get()`.

**2. Forgetting that the loop variable is a *key*.**
`for x in grades: print(x)` prints names, not scores. You need `grades[x]`.

**3. Using `in` and expecting it to search values.**
`88 in grades` is `False` even when Amina scored 88.

**4. Trying to index a dictionary by position.** `grades[0]` is not "the first
item" — it is a lookup for the key `0`, and that is another `KeyError`.

````python
grades = {"Amina": 88, "Ben": 72}
print(grades[0])
````

```
KeyError: 0
```

Python did **not** say "dictionaries are not indexable" — it went looking for a
key called `0`, found none, and raised `KeyError`. Dictionaries have no first,
second, or last item.

Compare: on a list, an out-of-range position gives `IndexError: list index out
of range`. **`KeyError` → dictionary. `IndexError` → list.**

## Summary

- A **dictionary** maps **keys → values**: `{"Ben": 72}`. Lookup by name, not by
  position. Written with `{}`, accessed with `d[key]`.
- `d[key] = value` **adds** if the key is new, **overwrites** if it is not.
  Keys are unique; the dictionary is mutable.
- A missing key raises **`KeyError`**. Guard with `key in d` or `d.get(key, default)`.
- **The counting pattern** — memorise it:
  `counts[item] = counts.get(item, 0) + 1`
- Loop with `for k in d`, `.keys()`, `.values()`, `.items()`. **Never rely on
  dictionary order** — sort it next week.
- Choose a dict when you look things up **by name**; a list when **order and
  position** carry the meaning. `[CLO 3]`
