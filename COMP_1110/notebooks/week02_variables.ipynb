## Values and types

A **value** is the most basic thing a program works with: a number, a piece of text.
Every value has a **type**, which decides what you may do with it.

| Type | Means | Examples |
| :--- | :--- | :--- |
| `int` | whole number | `17`, `0`, `-4` |
| `float` | number with a decimal point | `3.2`, `98.6`, `-0.5` |
| `str` | text ("string" of characters) | `"Nadia"`, `'B+'`, `"17"` |

Note the last one: `17` and `"17"` are **not** the same value.

````python
print(17)
print(3.2)
print("Nadia")

# The quotes are not part of the value - they tell Python "this is text".
print("17")
````

**Careful with commas.** You might write a big number the way you would on paper.
Python reads commas as separators, not decoration:

````python
print(1000000)
print(1, 000, 000)
````

### `type()` tells you what you have

When you are not sure what a value is, ask Python. `type()` is your first
debugging tool, and you will use it all term.

````python
print(type(17))
print(type(3.2))
print(type("Nadia"))
print(type("17"))
````

`<class 'int'>` is Python's way of saying "this is an integer".

Watch the last line: `"17"` has quotes, so it is a **`str`**, not an `int`.
That single fact causes more Week 2 bugs than anything else.

## Variables and assignment

A **variable** is a name that refers to a value. You create one with an
**assignment statement**:

```python
quiz_score = 18
```

Read `=` as **"gets"**, not "equals": *quiz_score gets the value 18.*
It is an instruction, not a claim of fact. Name on the **left**, value on the **right**.

````python
quiz_score = 18
course_name = "COMP 1110"
room_temperature = 21.5

print(course_name)
print(quiz_score)
print(room_temperature)
````

### A variable can be reassigned

Assignment is not permanent. A new assignment replaces the old value, and the
old value is simply gone.

````python
songs_played = 3
print(songs_played)

songs_played = songs_played + 1
print(songs_played)

songs_played = "many"
print(songs_played, type(songs_played))
````

Line 4 looks like nonsense in mathematics. In Python it is fine: **the right
side is worked out first** (`3 + 1` is `4`), then that result is stored back
into `songs_played`.

The last lines also show a variable's *type* can change. Legal — but usually a
sign of a confusing program.

## Variable names and keywords

Rules Python enforces:

- Letters, digits and underscores only — `total_marks`, `score2`.
- **Cannot start with a digit.** `2nd_score` is a syntax error.
- **No spaces, no hyphens.** `quiz score` and `quiz-score` are errors.
- **Case matters.** `Score` and `score` are two different variables.
- You cannot use one of Python's **keywords**.

### The keywords

```
and     as       assert   break    class   continue  def     del
elif    else     except   False    finally for       from    global
if      import   in       is       lambda  None      nonlocal not
or      pass     raise    return   True    try       while   with   yield
```

Do not memorise this. Your editor colours keywords differently — if a name you
typed changes colour, pick another name.

````python
# Legal names.
final_grade = 88
student_2 = "Ravi"
_hidden = True

print(final_grade, student_2, _hidden)
````

````python
# Illegal: a variable name cannot begin with a digit.
2nd_score = 91
````

## Statements

A **statement** is a unit of code Python can execute. You already know two kinds:

```python
print("hello")        # an expression statement - it produces output
midterm_mark = 74     # an assignment statement - it produces no output
```

A **script** is a sequence of statements run **top to bottom, one at a time**.
Assignment prints nothing — if you assign and see no output, that is not a bug.

````python
bus_fare = 2.75          # runs first, prints nothing
trips = 4                # runs second, prints nothing
print(bus_fare * trips)  # runs third, and this one prints
````

## Operators and operands

**Operators** are symbols that stand for a computation; **operands** are the
values they work on. In `price * quantity`, `*` is the operator and
`price`, `quantity` are the operands.

| Operator | Meaning | on `13` and `4` |
| :--- | :--- | :--- |
| `+` `-` `*` | add, subtract, multiply | `17`, `9`, `52` |
| `/` | division — **always a `float`** | `3.25` |
| `//` | floor division — drops the remainder | `3` |
| `%` | modulus — the remainder | `1` |
| `**` | power | `28561` |

````python
print(13 + 4)
print(13 - 4)
print(13 * 4)
print(13 / 4)
print(13 // 4)
print(13 % 4)
print(13 ** 4)
````

### `/` versus `//`

In Python 3, `/` **always** gives a `float`, even when the division comes out
exact. Use `//` when you want a whole number of things.

````python
print(8 / 2)     # 4.0 - a float, note the .0
print(8 // 2)    # 4   - an int
print(type(8 / 2))

# 100 minutes is how many whole hours, and how many minutes left over?
print(100 // 60, "hours and", 100 % 60, "minutes")
````

`%` is more useful than it looks: `n % 2` is `0` for every even number, and
`n % 10` gives you the last digit. We use it in Week 3 and Week 5.

## Expressions

An **expression** is any combination of values, variables and operators that
Python can reduce to a single value:

```python
17                 # an expression (a value on its own)
quiz_score         # an expression (a variable on its own)
quiz_score + 5     # an expression
```

Python **evaluates** it down to one value, then prints it or stores it.

````python
hours_worked = 12
hourly_pay = 17.40

gross_pay = hours_worked * hourly_pay
print(gross_pay)

# An expression can be built from variables you already computed.
take_home = gross_pay * 0.85
print(take_home)
````

## Order of operations

With several operators, Python follows **PEMDAS**:

1. **P**arentheses
2. **E**xponentiation `**`
3. **M**ultiplication and **D**ivision (`*`, `/`, `//`, `%`) — left to right
4. **A**ddition and **S**ubtraction — left to right

When in doubt, **add parentheses**. They cost nothing and they tell the next
reader what you meant.

````python
print(2 + 3 * 4)        # multiplication first
print((2 + 3) * 4)      # parentheses win
print(2 ** 3 ** 2)      # ** goes right to left: 2 ** 9
print(100 - 10 - 5)     # - goes left to right: 90 - 5
````

### A real one: averaging three quizzes

The parentheses here are not optional — they are the whole program.

````python
quiz_1 = 15
quiz_2 = 18
quiz_3 = 12

wrong_average = quiz_1 + quiz_2 + quiz_3 / 3
right_average = (quiz_1 + quiz_2 + quiz_3) / 3

print("Wrong:", wrong_average)
print("Right:", right_average)
````

## String operations

`+` and `*` work on strings too, but they mean something different:

- `+` **concatenates** — joins two strings end to end.
- `*` **repeats** a string a whole number of times.

`-` and `/` do not work on strings at all.

````python
first_name = "Amara"
last_name = "Okafor"

full_name = first_name + " " + last_name
print(full_name)

print("-" * 30)
print("ha" * 3)
````

Notice the `" "` in the middle: concatenation adds **nothing** of its own, so a
space between two words is yours to supply.

### `print` with commas versus `+`

Two ways to print several things — and they are not the same.

````python
city = "Kamloops"
high_temp = 27

# Commas: print puts a space between items and accepts any type.
print(city, "high today:", high_temp)

# Plus: joins strings only - every piece must already be a string.
print(city + " high today: " + str(high_temp))
````

## Predict the output (3)

```python
print("3" + "4")
print(3 + 4)
print("3" * 4)
```

Three lines. What comes out?

**Answer**

```
34
7
3333
```

- `"3" + "4"` — both strings, so `+` **joins** them into the text `34`.
- `3 + 4` — both ints, so `+` **adds**: `7`.
- `"3" * 4` — string times int **repeats** the string four times.

Same operators, different types, completely different answers. **Types matter.**

## `input()` always gives you a `str`

This is the single most important sentence of Week 2.

Even when the user types `42`, what comes back is the **text** `"42"`, not the
number `42`. Python does not guess what you meant.

````python
# raw_age = input("How old are you? ")   <- the real line
raw_age = "42"        # imagine the user typed 42

print(raw_age)
print(type(raw_age))

# So this does NOT do arithmetic:
print(raw_age + "1")
````

## Type conversion: `int()`, `float()`, `str()`

| Call | Gives you | Example |
| :--- | :--- | :--- |
| `int(x)` | a whole number | `int("42")` → `42` |
| `float(x)` | a decimal number | `float("3.5")` → `3.5` |
| `str(x)` | text | `str(42)` → `"42"` |

The rule: **convert as soon as the value arrives**, then work with real numbers
for the rest of the program.

````python
raw_age = "42"           # what input() would have handed you

age = int(raw_age)       # convert once, right away
print(age, type(age))

print("Next year you will be", age + 1)
````

````python
print(int("42"))         # str  -> int
print(float("3.5"))      # str  -> float
print(str(42) + "!")     # int  -> str, so + can join it
print(int(3.9))          # float-> int: chops off, does NOT round
print(float(7))          # int  -> float
````

### The classic bug: `"3" + 4`

Half the class will hit this in Worksheet 2. Here it is on purpose.

````python
# quantity = input("How many coffees? ")   <- the real line
quantity = "3"          # imagine the user typed 3

total_cups = quantity + 4
print(total_cups)
````

```
TypeError: can only concatenate str (not "int") to str
```

Python is saying: *the left side is text, so `+` means "join" — and I cannot
join a number onto text.* **The fix is one call:** convert before you compute.

````python
quantity = "3"          # imagine the user typed 3

total_cups = int(quantity) + 4
print(total_cups)
````

### Conversion can fail too

`int()` needs text that really is a whole number.

````python
print(int("42"))       # fine
print(int(" 42 "))     # fine - surrounding spaces are ignored
print(int("forty"))    # not fine
````

`ValueError: invalid literal for int() with base 10: 'forty'` means
"you promised me a number and gave me a word."

Handling that gracefully instead of crashing is `try` / `except` — **next week**.

## Comments

Anything after a `#` on a line is ignored by Python. It is there for **people**.

```python
# Convert the raw input to a number before doing arithmetic.
age = int(raw_age)
```

Write comments that explain **why**, not **what**. The code already says what it
does; it cannot say why you chose to do it.

````python
# BAD - just re-reads the code out loud.
seconds = 7200      # set seconds to 7200

# GOOD - explains the reason and the units.
seconds = 7200      # the two-hour lab booking, in seconds

hours = seconds // 3600
print("Booking length:", hours, "hours")
````

## Choosing mnemonic variable names

*Mnemonic* means **memory-aiding**. A good name reminds the reader — usually
you, three weeks later — what the value is for.

```python
x = 0.13 * 24.99          # what on earth is this?
tax = 0.13 * item_price   # oh, it is sales tax
```

Python does not care; it runs both. **Your marker cares** — worksheets are
graded for readable code as well as correct output.

````python
# Hard to read - and just as hard to debug.
a = 24.99
b = 3
c = a * b
print(c)
````

````python
# Same arithmetic, but the program now explains itself.
item_price = 24.99
item_count = 3

subtotal = item_price * item_count
print("Subtotal:", subtotal)
````

### Habits to start now

- Use whole words: `student_count`, not `sc` or `n1`.
- Separate words with underscores: `final_grade`. This is called **snake_case**.
- Put spaces around operators: `total = price * quantity`, not `total=price*quantity`.
- Avoid single letters except for genuine counters (`i` in a loop, from Week 5).

Python has a written style guide and we spend all of **Week 13** on it. For now
just build the habits — you will get the rules for free later.

## Debugging

Three kinds of error, and they fail in different ways:

| Kind | When it bites | Example |
| :--- | :--- | :--- |
| **Syntax** | before anything runs | `print("hi"` — missing bracket |
| **Runtime** | part-way through | `int("forty")` — `ValueError` |
| **Semantic** | never — it just gives a wrong answer | `q1 + q2 + q3 / 3` |

The third is the dangerous one: Python is perfectly happy, and only you can
notice the answer is wrong.

### A method that works

1. **Read the error from the bottom up.** The last line names the error type;
   the line above gives the line number.
2. **Print the variable and its type** just before the line that failed:
   `print(quantity, type(quantity))`.
3. **Cut the program in half.** Comment out the bottom — does the top still work?
4. **Say the line out loud.** "quantity gets the string three." Often that is enough.
5. **Take a break.** Ten minutes away beats an hour of staring.

## Common errors this week

**1. Adding a string to a number**

```python
total = quantity + 4      # quantity came from input()
```
```
TypeError: can only concatenate str (not "int") to str
```
*Meaning:* the left operand is text, so `+` tried to join rather than add.
**Fix:** `int(quantity) + 4`.

**2. Using a variable you never assigned (or misspelled)**

````python
student_name = "Priya"
print(studnet_name)
````

`NameError: name 'studnet_name' is not defined` — Python has never seen that
name. **Fix:** check the spelling *and* the capitals; `Name` and `name` differ.

**3. Assigning in the wrong direction**

````python
17 = quiz_score
````

`SyntaxError: cannot assign to literal here.` — the name must be on the
**left** of the `=`. **Fix:** `quiz_score = 17`.

**4. The one Python will not warn you about**

```python
average = quiz_1 + quiz_2 + quiz_3 / 3
```

No error message. No red text. Just a wrong number, silently, forever.
**Fix:** `(quiz_1 + quiz_2 + quiz_3) / 3` — and always check one answer by hand.

## Summary

- Every value has a **type**: `int`, `float`, `str`. Ask `type()` when unsure.
- `=` means **"gets"**: the right side is evaluated first, then stored in the
  name on the left.
- Operators `+ - * **`, plus `/` (always a float), `//` (whole), `%` (remainder).
  Parentheses beat everything.
- On strings, `+` **joins** and `*` **repeats** — same symbol, different job.
- **`input()` always returns a `str`** — convert with `int()` or `float()`
  immediately, or you get a `TypeError`.
- Mnemonic names and *why* comments are part of a correct answer, not decoration.
