## A boolean is a yes/no answer

Every decision a program makes comes down to one question:

> Is this **True**, or is it **False**?

`True` and `False` are values, just like `7` and `"hello"` are values.
They have their own type: **`bool`**.

Capitalisation matters. `True` is a value; `true` is an undefined name.

````python
is_raining = True
is_weekend = False

print(is_raining)
print(type(is_raining))
print(is_weekend, type(is_weekend))
````

## Comparison operators produce booleans

You rarely type `True` yourself. You **compute** it by comparing two values.

| Operator | Meaning |
| :--- | :--- |
| `x == y` | x is equal to y |
| `x != y` | x is not equal to y |
| `x > y` | x is greater than y |
| `x < y` | x is less than y |
| `x >= y` | x is greater than or equal to y |
| `x <= y` | x is less than or equal to y |

Each of these is an **expression**: it evaluates to `True` or `False`.

````python
score = 78
passing_mark = 50

print(score > passing_mark)
print(score == 100)
print(score != 78)
print(score <= 78)

# Comparisons work on strings too (alphabetical order).
print("apple" < "banana")
````

### The single most common beginner error

- `=` is **assignment**. It stores a value: `score = 78`
- `==` is **comparison**. It asks a question: `score == 78`

Read `=` out loud as *"gets"* and `==` as *"is equal to"*.
`score = 78` is "score **gets** 78". `score == 78` is "**is** score equal to 78?"

Python will not let you use `=` where a question belongs — you get a
`SyntaxError`. We will see that error in the Common Errors slide.

## Combining questions: `and`, `or`, `not`

Real conditions have more than one part:

> "Is the temperature above 15 **and** is it not raining?"

Python spells these out as words, not symbols:

- `and` — True only when **both** sides are True
- `or` — True when **at least one** side is True
- `not` — flips True to False and False to True

````python
temperature = 22
is_raining = False

print(temperature > 15 and not is_raining)   # good day for a walk
print(temperature > 30 or temperature < 0)   # extreme weather?
print(not is_raining)
````

### Truth tables

| `a` | `b` | `a and b` | `a or b` |
| :--- | :--- | :--- | :--- |
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

| `a` | `not a` |
| :--- | :--- |
| True | False |
| False | True |

`or` in Python is **inclusive**: "both" also counts as True.
Everyday English often means "one or the other, not both" — Python never does.

````python
age = 19
has_ticket = True
is_member = False

print(age >= 18 and has_ticket)          # True  and True  -> True
print(is_member or has_ticket)           # False or True   -> True
print(age >= 18 and is_member)           # True  and False -> False
print(not (age >= 18))                   # not True        -> False
````

## `if`: do this only when the condition holds

The simplest conditional. One question, one block of code that runs
**only if the answer is True**.

```python
if <boolean expression>:
    <body — runs only when the expression is True>
```

Two pieces of punctuation carry the whole meaning:

1. the **colon** `:` at the end of the `if` line
2. the **indentation** of the lines beneath it

````python
temperature = 33

if temperature > 30:
    print("Heat warning.")
    print("Drink water and stay in the shade.")

print("Have a good day.")   # not indented -> always runs
````

## Indentation is syntax, not decoration

In many languages, indentation is a courtesy to human readers.
**In Python it is part of the grammar.** The indentation *is* how you tell
Python where a block begins and ends.

- Use **4 spaces** per level. Always. (PEP 8, which we cover in Week 13.)
- Every line in the same block must be indented the **same amount**.
- Configure your editor to insert spaces when you press Tab. Never mix
  tabs and spaces in one file.

Change the indentation, and you change what the program **means** —
not just how it looks.

````python
mark = 91

if mark >= 90:
    print("Excellent work!")
    print("This line is INSIDE the if.")

print("This line is OUTSIDE the if.")
````

````python
mark = 42   # now the condition is False

if mark >= 90:
    print("Excellent work!")
    print("This line is INSIDE the if.")

print("This line is OUTSIDE the if.")
````

### When the indentation is wrong

Python does not guess. It stops and tells you.

The cell below is missing the indented body after `if`. Run it and read the
message — you will meet it many times this term.

````python
mark = 91

if mark >= 90:
print("Excellent work!")
````

`IndentationError: expected an indented block after 'if' statement on line 3`

Read it bottom-up, as the syllabus advises:

- the **last** line names the error: `IndentationError`
- it tells you **what** was expected: an indented block
- and **where**: after the `if` on line 3

The fix is always the same: indent the body by 4 spaces.

## `if` / `else`: exactly one of two paths

When there is a sensible thing to do in *both* cases, add `else`.

```python
if <condition>:
    <runs when True>
else:
    <runs when False>
```

Exactly one branch runs. Never both, never neither.
`else` takes no condition of its own — it means "in every other case".

````python
mark = 47

if mark >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL")

print("Mark recorded:", mark)
````

## `elif`: more than two paths

A letter grade has five outcomes, not two. Chain them with `elif`
("else if"):

- Python checks conditions **top to bottom**.
- The **first** True branch runs; everything below it is skipped.
- A final `else` catches everything that matched nothing.

You may have as many `elif` branches as you need, and the `else` is optional.

````python
mark = 84

if mark >= 90:
    grade = "A"
elif mark >= 80:
    grade = "B"
elif mark >= 70:
    grade = "C"
elif mark >= 50:
    grade = "D"
else:
    grade = "F"

print("Mark", mark, "earns grade", grade)
````

### Order matters — a lot

Because the **first** True branch wins, `elif` conditions must go from most
specific to least specific.

Write the grade chain the other way round and every passing mark becomes a `D`:
`84 >= 50` is True, so Python stops there and never looks at the `B` branch.

````python
mark = 84

# BUG: conditions are in the wrong order.
if mark >= 50:
    grade = "D"
elif mark >= 80:
    grade = "B"
else:
    grade = "F"

print("Reported grade:", grade, "- should have been B")
````

## Nested conditionals

A branch's body can contain another `if`. Each level of nesting adds 4 more
spaces of indentation.

Use it when the second question only makes sense **after** the first one is
answered — for example, only check the shipping option if there is an order
at all.

Nesting deeper than two levels is usually a sign the logic can be simplified.

````python
order_total = 75
is_member = True

if order_total > 0:
    if is_member:
        print("Member discount applied.")
        print("You pay:", order_total * 0.9)
    else:
        print("You pay:", order_total)
else:
    print("Your cart is empty.")
````

### Flatten it with `and` when you can

The nested version above tests two things in sequence. When both tests are
independent, one `and` reads better than two levels of indentation:

Fewer levels, less indentation to get wrong, same behaviour.

````python
order_total = 75
is_member = True

if order_total > 0 and is_member:
    print("You pay:", order_total * 0.9)
elif order_total > 0:
    print("You pay:", order_total)
else:
    print("Your cart is empty.")
````

## When good input goes bad

Last week we converted text to a number:

```python
age = int(input("Your age: "))
```

That works — right up until someone types `twenty`, or their name, or just
presses Enter.

`int()` cannot convert that. It raises a **`ValueError`**, and an uncaught
exception **stops your program dead**. Everything after it never runs.

````python
# Imagine the user typed this at an input() prompt:
typed_by_user = "twenty"

age = int(typed_by_user)
print("This line never runs.")
````

`ValueError: invalid literal for int() with base 10: 'twenty'`

Note the last line: the program **crashed**. `print("This line never runs.")`
was never reached.

We cannot stop users typing nonsense. We can stop it from killing the program.

## `try` / `except`

Wrap the risky code in `try`. If it raises an exception, Python abandons the
rest of the `try` block and runs `except` instead.

```python
try:
    <risky code>
except:
    <what to do when it goes wrong>
```

If nothing goes wrong, the `except` block is skipped entirely.
This is how a program **recovers** instead of crashing.

````python
typed_by_user = "twenty"   # imagine input() returned this

try:
    age = int(typed_by_user)
    print("Next year you will be", age + 1)
except:
    age = -1
    print("That was not a whole number. Please type digits only.")

print("The program is still running.")
````

### The same code with valid input

Nothing raises, so the `except` block never runs.

````python
typed_by_user = "24"

try:
    age = int(typed_by_user)
    print("Next year you will be", age + 1)
except:
    age = -1
    print("That was not a whole number. Please type digits only.")

print("The program is still running.")
````

### The standard input-guarding pattern

In a real program you would write:

```python
raw = input("Enter your age: ")
try:
    age = int(raw)
except:
    print("Please enter a whole number.")
    age = 0
```

Two rules of thumb:

- Put **as little as possible** inside `try` — only the line that can fail.
- Always leave the variable in a usable state in `except`, or the code
  below it will fail next.

## Short-circuit evaluation

Python is lazy about `and` and `or`. It evaluates left to right and stops as
soon as the answer is settled:

- `False and <anything>` → already `False`. The right side is **never evaluated**.
- `True or <anything>` → already `True`. The right side is **never evaluated**.

This is not a curiosity — it is how you protect a dangerous operation.

````python
count = 0

# Dividing by zero raises ZeroDivisionError. The left test prevents it:
print(count != 0 and 100 / count > 10)

count = 5
print(count != 0 and 100 / count > 10)   # 20.0 > 10 -> True
````

### The guardian pattern

Put the **guard** on the left, the risky expression on the right:

```python
if count != 0 and 100 / count > 10:
```

Swap the two sides and the program crashes with `ZeroDivisionError` when
`count` is 0, because Python evaluates the division before it ever reaches
the guard.

Same idea, `or` version: `if count == 0 or 100 / count > 10:`

````python
temperature_text = ""   # user pressed Enter without typing

# Guard first: if the text is empty, never attempt the conversion.
if temperature_text != "" and int(temperature_text) > 30:
    print("Hot!")
else:
    print("No usable temperature reading.")
````

## Common errors — 1

**`=` used where `==` belongs**

```python
if mark = 50:
    print("Exactly half")
```

> `SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?`

**Means:** you told Python to *assign* inside a question. A condition must be
an expression that produces `True` or `False`; assignment produces nothing.
**Fix:** `if mark == 50:`

## Common errors — 2: punctuation and indentation

```python
if mark >= 50        # missing colon
    print("Pass")
```
> `SyntaxError: expected ':'` — Python reached the end of the `if` line and
> found no `:` to open the block. Same for `elif`, `else`, `try`, `except`.

```python
if mark >= 50:
    print("Pass")
      print("Well done")   # jumped in further for no reason
```
> `IndentationError: unexpected indent` — lines in one block must be indented
> **identically**. 4 spaces per level, never tabs mixed with spaces.

## Common errors — 3

**`elif` conditions in the wrong order**

```python
if mark >= 50:
    grade = "D"
elif mark >= 80:
    grade = "B"
```

> No error message at all — it just gives the wrong answer.

**Means:** a **logic error**. The first True branch wins, and `mark >= 50` is
True for every B student. **Fix:** order conditions from most restrictive to
least. Test with a value from every branch.

## Summary

1. A **boolean expression** evaluates to `True` or `False`. Comparison
   operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) produce them; `and`, `or`,
   `not` combine them.
2. `=` assigns, `==` compares. Mixing them up is a `SyntaxError`.
3. `if` / `else` chooses one of two paths; an `elif` chain chooses one of
   many, and the **first** True branch wins — so order matters.
4. **Indentation is syntax.** 4 spaces per level, consistent within a block.
   Wrong indentation is either an `IndentationError` or a silent logic bug.
5. `try` / `except` lets a program survive bad input instead of crashing on
   `int()`.
6. `and` / `or` **short-circuit** — put the guard on the left.
