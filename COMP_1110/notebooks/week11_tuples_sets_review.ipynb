# Part 1 — Tuples

## A tuple is a sequence you promise not to change

- Like a list: ordered, indexed from `0`, can hold any types, can be looped over.
- Unlike a list: **immutable** — once built, it cannot be modified.

Why would we ever want *less* power? Because "this will not change" is useful:
a date `(2026, 11, 17)`, a point `(x, y)`, a person `("Amara", 19)`.
Immutability makes tuples safe to hand around, and lets them be **dictionary keys**.

**Creating one:** round brackets — or often just the commas.

````python
student = ("Amara", 19, "Computing")
print(student)
print(student[0])
print(len(student))

# The brackets are optional; the commas are what make it a tuple.
temperatures = 4, -2, 11
print(temperatures)
print(type(temperatures))
````

**Watch the one-element tuple:** the comma, not the bracket, does the work.

````python
not_a_tuple = ("Amara")
one_tuple = ("Amara",)
print(type(not_a_tuple))
print(type(one_tuple))
print(len(one_tuple))
````

## Tuples are immutable

A list lets you assign to an element. A tuple does not.

```python
prices = [4.50, 2.25]
prices[0] = 5.00      # fine — lists are mutable

sizes = (4.50, 2.25)
sizes[0] = 5.00       # TypeError — tuples are immutable
```

To "change" a tuple you build a **new** one.

````python
sizes = (4.50, 2.25)
sizes[0] = 5.00
````

`TypeError: 'tuple' object does not support item assignment` —
Python is telling you the object itself refuses the operation.
Build a new tuple instead:

````python
sizes = (4.50, 2.25)
sizes = (5.00,) + sizes[1:]
print(sizes)
````

## Comparing tuples

Python compares tuples **element by element, left to right** — the first difference decides.
This is exactly how words are ordered in a dictionary, one letter at a time.

That single rule is why sorting a **list of tuples** just works.

````python
print((0, 1, 2) < (0, 3, 4))     # 0 == 0, then 1 < 3 -> True
print((1, 100) < (2, 1))         # first element decides -> True
print(("Amara", 19) < ("Amara", 21))

scores = [(88, "Priya"), (72, "Dev"), (95, "Lin"), (88, "Ana")]
scores.sort()
print(scores)
````

## Tuple assignment and multiple assignment

The left-hand side can be a *tuple of names*. Python unpacks position by position.

- Swap two variables in one line — no temporary variable.
- Split a string and unpack the pieces straight into named variables.
- The number of names must match the number of values, or you get a `ValueError`.

````python
first = "salt"
second = "pepper"
first, second = second, first
print(first, second)

email = "amara@tru.ca"
user, domain = email.split("@")
print(user)
print(domain)
````

## Tuples as return values

A function can only return one object — but that object can be a tuple,
so effectively a function can return **several values at once**.

The caller unpacks them with tuple assignment.

````python
def min_and_max(numbers):
    """Return the smallest and largest value in a list of numbers."""
    return min(numbers), max(numbers)


temps = [4, -2, 11, 7, -5]
coldest, warmest = min_and_max(temps)
print("Coldest:", coldest)
print("Warmest:", warmest)
print(min_and_max(temps))
````

## Dictionaries and tuples: `.items()`

A dictionary is a bag of key/value **pairs** — and a pair is naturally a **tuple**.

`d.items()` gives you those pairs. Loop over it and unpack with tuple assignment:

```python
for key, value in d.items():
```

This is the cleanest way to loop over a dictionary when you need both halves.

````python
plays = {"Sunflower": 12, "Levitating": 47, "Anti-Hero": 30, "Bad Habit": 5}

print(list(plays.items()))

for song, count in plays.items():
    print(song, "->", count)
````

## The classic problem: sort a dictionary **by value**

Dictionaries have no order you control, and `sorted(plays)` sorts the **keys**.

We want the **most played songs first**. The trick, in three steps:

1. Get the pairs with `.items()`.
2. Build a list of `(value, key)` tuples — value **first**.
3. `sorted()` compares tuples left to right, so it sorts by value automatically.

Learn this pattern. It shows up on exams, on worksheets, and in your project.

### Step 1 and 2 — get the pairs, then flip them

````python
plays = {"Sunflower": 12, "Levitating": 47, "Anti-Hero": 30, "Bad Habit": 5}

pairs = []
for song, count in plays.items():
    pairs.append((count, song))    # value FIRST, key second

print(pairs)
````

### Step 3 — sort, reverse, report

````python
ordered = sorted(pairs)             # ascending, by count then by song
print(ordered)

ordered.reverse()                   # most played first
print(ordered)

print("Top 3:")
for count, song in ordered[:3]:
    print(song, count)
````

**The whole pattern, compressed** — worth memorising the shape:

```python
pairs = []
for key, value in d.items():
    pairs.append((value, key))
pairs.sort()
pairs.reverse()
```

`sorted(pairs, reverse=True)` does the sort and the flip in one call.

## Tuples as dictionary keys

Lists **cannot** be dictionary keys — they are mutable, so their identity could change
underneath the dictionary. Tuples **can**, because they never change.

So when the thing you are counting has **two parts**, use a tuple key:
`(first_name, last_name)`, `(city, month)`, `(row, column)`.

````python
sales = [("Kamloops", "Nov"), ("Vernon", "Nov"),
         ("Kamloops", "Dec"), ("Kamloops", "Nov")]

counts = {}
for pair in sales:
    counts[pair] = counts.get(pair, 0) + 1

for location, month in counts:
    print(location, month, counts[(location, month)])
````

## The three sequences, side by side

| | string | list | tuple |
| :--- | :--- | :--- | :--- |
| written as | `"hi"` | `[1, 2]` | `(1, 2)` |
| ordered / indexable | yes | yes | yes |
| **mutable** | **no** | **yes** | **no** |
| can be a dict key | yes | **no** | yes |
| holds | characters only | anything | anything |
| typical use | text | a collection you edit | a fixed record / a pair |

All three support `len()`, `in`, slicing, `for`, and `+`.

# Sets — the topic with no chapter

> **Not in *Python for Everybody*.** Study these notes plus the
> **supplementary set notes on Moodle**. Sets *are* examinable on Thursday.

A **set** is an unordered collection of **unique** items.

- No duplicates — ever. Adding something twice changes nothing.
- No order, no indexing — `my_set[0]` is an error.
- Extremely fast at answering one question: **"is this in here?"**

````python
majors = {"Computing", "Biology", "Computing", "Nursing"}
print(majors)
print(len(majors))

# From a list — duplicates silently disappear
votes = ["yes", "no", "yes", "yes", "no"]
print(set(votes))

empty = set()          # NOT {} — that is an empty dictionary
print(empty, type(empty))
````

## Adding, removing, and asking "is it in here?"

- `add(item)` — put one item in (no effect if already present).
- `discard(item)` — take it out, **no error** if it was never there.
- `remove(item)` — take it out, `KeyError` if it was never there.
- `item in my_set` — the question sets are built for.

For a list, `in` checks items one at a time. For a set, Python jumps almost straight
to the answer — so on big collections a set is dramatically faster.

````python
seen_ids = set()
seen_ids.add("T00123")
seen_ids.add("T00456")
seen_ids.add("T00123")        # already there — no change, no error
print(seen_ids, len(seen_ids))

seen_ids.discard("T00999")    # not present — silently fine
print("T00456" in seen_ids)
print("T00999" in seen_ids)
````

````python
import time

big_list = list(range(200000))
big_set = set(big_list)

start = time.perf_counter()
print(199999 in big_list)
list_time = time.perf_counter() - start

start = time.perf_counter()
print(199999 in big_set)
set_time = time.perf_counter() - start

print("list search took about", round(list_time * 1000, 3), "ms")
print("set  search took about", round(set_time * 1000, 3), "ms")
````

## Set operations: union, intersection, difference

Given two sets of students:

| operation | operator | method | meaning |
| :--- | :--- | :--- | :--- |
| union | `a \| b` | `a.union(b)` | in **either** set |
| intersection | `a & b` | `a.intersection(b)` | in **both** sets |
| difference | `a - b` | `a.difference(b)` | in `a` but **not** in `b` |

These answer real questions in one line: who is enrolled in both courses? who dropped?

````python
comp1110 = {"Amara", "Dev", "Lin", "Priya"}
math1130 = {"Dev", "Priya", "Sam"}

print(comp1110 | math1130)      # everyone taking at least one
print(comp1110 & math1130)      # taking both
print(comp1110 - math1130)      # only in COMP 1110
print(math1130 - comp1110)      # only in MATH 1130
````

## When is a set the right choice?

Reach for a set when you need one of these:

1. **Deduplication** — "give me the distinct words / student IDs / cities".
   `list(set(words))` is the whole job.
2. **"Have I seen this before?"** — keep a `seen` set, test with `in`, `add` as you go.
3. **Comparing two groups** — who is in both, who is only in one (`&`, `-`, `|`).

Do **not** use a set when you need **order**, **duplicates**, **indexing**,
or **counts** — those are jobs for a list or a dictionary.

````python
# LIVE CODING
# Given: text = "the cat sat on the mat with the cat"
# 1. Report the first word that appears a second time.
# 2. Report how many DISTINCT words the line contains.
# Hint: a `seen` set, a for loop, and `in`.

text = "the cat sat on the mat with the cat"
````

````python
# Solution
text = "the cat sat on the mat with the cat"

seen = set()
first_repeat = None
for word in text.split():
    if word in seen and first_repeat is None:
        first_repeat = word
    seen.add(word)

print("First repeated word:", first_repeat)
print("Distinct words:", len(seen))
````

## Common errors — tuples and sets

| What you write | What Python says | What it means |
| :--- | :--- | :--- |
| `point[0] = 5` on a tuple | `TypeError: 'tuple' object does not support item assignment` | Tuples never change — build a new one. |
| `t = ("Amara")` | *no error*, but `type(t)` is `str` | A one-item tuple needs the comma: `("Amara",)`. |
| `s = {}` then `s.add(1)` | `AttributeError: 'dict' object has no attribute 'add'` | `{}` is an empty **dict**. Use `set()`. |
| `my_set[0]` | `TypeError: 'set' object is not subscriptable` | Sets have no order and no index. Loop, or convert to a list. |
| `a, b = "x y z".split()` | `ValueError: too many values to unpack` | Names on the left must match values on the right. |

### Exam-style: find the bug

This should print each student's name and their mark plus 5, then the best
possible mark. It prints the three students fine — and then crashes. Why?

```python
handle = open("week11_grades.txt")
for line in handle:
    name, mark = line.split()
    print(name, int(mark) + 5)
print("Best possible:", mark + 5)
```

### Answer

**`mark` is still a string.** Inside the loop the conversion `int(mark)` was thrown away —
its result was printed, never stored. So after the loop `mark` is `"95"`, and
`mark + 5` raises `TypeError: can only concatenate str (not "int") to str`.

Two habits this teaches:

- **Convert once and keep it**: `mark = int(mark)` inside the loop.
- A loop variable **survives** after the loop, still holding the **last** value.

(`split()` happens to discard the `\n` for you here. Printing `line` directly without
`rstrip()` is what gives you the double-spaced output.)

````python
handle = open("week11_grades.txt")
best = 0
for line in handle:
    name, mark = line.split()
    mark = int(mark)
    if mark > best:
        best = mark
    print(name, mark + 5)
handle.close()
print("Best possible:", best + 5)
````

## Ch. 8 — Lists, in one slide

- **Mutable**: `nums[0] = 9`, `nums.append(4)`, `nums.sort()` change the list **in place**
  and return `None`. `x = nums.sort()` sets `x` to `None` — a classic exam trap.
- `sorted(nums)` returns a **new** sorted list and leaves the original alone.
- `"a,b,c".split(",")` → list; `",".join(["a", "b"])` → string.
- **Aliasing:** `b = a` makes a second name for the *same* list. `b = a[:]` copies it.
- A list passed into a function can be modified by that function.

### Exam-style: trace the aliasing

```python

b = a
c = a[:]
b.append(4)
c.append(5)
print(a)
print(c)
print(sorted([3, 1, 2]) == [1, 2, 3])
print([3, 1, 2].sort())
```

### Answer

```
[1, 2, 3, 4]
[1, 2, 3, 5]
True
None
```

`b` **is** `a` — appending through `b` changes `a`. `c` is a copy, so it goes its own way.
`.sort()` returns `None`; `sorted()` returns the list.

## Ch. 9 — Dictionaries, in one slide

- `d[key]` on a missing key → `KeyError`. `d.get(key, 0)` returns a default instead.
- **The counting pattern** — know it cold:

```python
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
```

- `for key in d:` loops over **keys**; `for key, value in d.items():` gives both.
- Keys must be immutable: strings, numbers, **tuples** — never lists.

### Exam-style: trace the counting

```python
counts = {}
for word in "the cat the hat the end".split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
print(len(counts))
print(counts.get("dog", 0))
print(sorted(counts))
```

### Answer

```
{'the': 3, 'cat': 1, 'hat': 1, 'end': 1}
4
0
['cat', 'end', 'hat', 'the']
```

`len` counts **keys**, not occurrences. `get` with a default never raises.
`sorted(d)` sorts the **keys** — to sort by value you need the `(value, key)` tuple trick.

````python
counts = {}
for word in "the cat the hat the end".split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
print(len(counts))
print(counts.get("dog", 0))
print(sorted(counts))
````

## Week 8 — testing and documentation

- A **test plan** lists inputs and the expected output *before* you write the code.
- **Edge cases** are where marks live: empty list, empty string, zero, negatives,
  one element, missing key, missing file.
- `assert condition, "message"` — passes silently, raises `AssertionError` when false.
- `unittest`: a class extending `unittest.TestCase`, methods starting with `test_`,
  checks like `assertEqual`. (Not responsible for `setUp`/`tearDown`.)
- A **docstring** is the `"""..."""` string on the *first line inside* a function:
  what it takes, what it returns. A `#` comment above the `def` is **not** a docstring.

### Exam-style: find the bug, then name the missing test

```python
def average(numbers):
    total = 0
    for n in numbers:
        total = total + n
    return total / len(numbers)

assert average([2, 4, 6]) == 4
assert average([]) == 0
```

What happens, and what would you change?

### Answer

The first `assert` passes. The second raises **`ZeroDivisionError`** — not an
`AssertionError` — because `len([])` is `0`, so the crash happens *inside* `average`
before the comparison ever runs.

The missing **edge case** is the empty list. Guard it:

```python
if len(numbers) == 0:
    return 0
```

Exam habit: for every function you write, ask "what does this do with **nothing**?"

````python
def average(numbers):
    """Return the mean of a list of numbers, or 0.0 for an empty list."""
    if len(numbers) == 0:
        return 0.0
    total = 0
    for n in numbers:
        total = total + n
    return total / len(numbers)


assert average([2, 4, 6]) == 4
assert average([]) == 0
print("All tests passed.")
print(average.__doc__)
````
