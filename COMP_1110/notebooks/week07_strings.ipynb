## A string is a *sequence* of characters

A string is not one indivisible blob. It is an ordered row of characters, each with a numbered
position — its **index**.

```
 s = "Python"

   P    y    t    h    o    n
   0    1    2    3    4    5
  -6   -5   -4   -3   -2   -1
```

Use square brackets to pull one character out: `s[0]`.

**Counting starts at 0.** The first character is at index `0`, not `1`.

````python
course = "Python"

print(course[0])
print(course[1])
print(course[5])
````

### Negative indexes count from the right

`s[-1]` is the **last** character. This is handy because you do not need to know the length.

````python
course = "Python"

print(course[-1])   # last character
print(course[-2])   # second-last
print(course[-6])   # same as course[0]
````

## `len()` — how long is it?

`len(s)` gives the number of characters in the string.

The **last valid index is `len(s) - 1`**, because indexes start at 0. Forgetting the `- 1` is the
single most common string bug in this course.

````python
message = "Good morning"

print(len(message))          # 12 characters (the space counts!)
print(message[len(message) - 1])   # last character
````

````python
message = "Good morning"
print(message[len(message)])   # one past the end
````

`IndexError: string index out of range` — index 12 does not exist in a 12-character string.
Valid indexes are `0` through `11`.

**Rule:** a slice `s[a:b]` has `b - a` characters when both are in range.

## Traversal: visiting every character

A very common pattern: do something once *per character*.

Two ways —

1. **With an index and a `while` loop** — you know *where* you are.
2. **With a `for` loop** — shorter, and you cannot fall off the end.

````python
fruit = "banana"

index = 0
while index < len(fruit):
    print(index, fruit[index])
    index = index + 1
````

### The same traversal with `for`

No counter, no `len`, no off-by-one risk. Prefer this when you do not need the index.

````python
fruit = "banana"

for letter in fruit:
    print(letter)
````

## Slices: taking a piece

`s[start:end]` gives the characters from `start` **up to but not including** `end`.

| Slice | Meaning |
| :--- | :--- |
| `s[2:5]` | indexes 2, 3, 4 |
| `s[:5]` | from the beginning to index 4 |
| `s[5:]` | from index 5 to the end |
| `s[:]` | the whole string |
| `s[-3:]` | the last three characters |

If `start >= end`, you get the **empty string** `''` — not an error.

````python
date_text = "2026-10-20"

print(date_text[0:4])    # year
print(date_text[5:7])    # month
print(date_text[8:])     # day
print(date_text[-2:])    # day again, counting from the right
````

````python
date_text = "2026-10-20"

print(repr(date_text[7:3]))    # start after end -> empty
print(repr(date_text[4:200]))  # end past the end is fine, no error
````

Slices are **forgiving**; single indexes are **not**. `date_text[200]` raises `IndexError`,
but `date_text[4:200]` just stops at the end of the string.

## Strings are **immutable**

You cannot change a character of an existing string. Once a string object exists, its characters
are fixed forever.

```python
greeting = "Hello, world"
greeting[0] = "J"      # ← TypeError
```

This is not a limitation to work around — it is a guarantee. If you pass a string to a function,
that function cannot secretly alter it.

````python
greeting = "Hello, world"
greeting[0] = "J"
````

`TypeError: 'str' object does not support item assignment`

**The fix: build a new string** instead of editing the old one.

````python
greeting = "Hello, world"
new_greeting = "J" + greeting[1:]

print(new_greeting)
print(greeting)          # the original is untouched
````

Every string "change" in Python is really *making a new string*. `upper()`, `strip()`, `replace()`
all **return** a new string and leave the original alone.

## Looping and counting

The counter pattern from Week 5, applied to text:

1. Set a counter to `0` **before** the loop.
2. Look at each character.
3. Add `1` when it matches.
4. Report **after** the loop.

This is the shape of almost every "how many …?" question about text.

````python
tweet = "the rain in spain stays mainly in the plain"

count = 0
for letter in tweet:
    if letter == "a":
        count = count + 1

print("Number of a's:", count)
````

### Wrap it in a function — reusable and testable

Week 4 said: if you did it twice, make it a function.

````python
def count_letter(text, target):
    """Return how many times target appears in text."""
    count = 0
    for letter in text:
        if letter == target:
            count = count + 1
    return count


print(count_letter("mississippi", "s"))
print(count_letter("mississippi", "z"))
````

## The `in` operator

`in` is a boolean operator that asks: *is this string a **substring** of that one?*

```python
"a" in "banana"        # True
"seed" in "banana"     # False
"nan" in "banana"      # True  -- substrings, not just single characters
```

It returns `True` or `False`, so it goes straight into an `if`.

````python
subject = "URGENT: your invoice is overdue"

if "urgent" in subject.lower():
    print("Flagging this message.")
else:
    print("Normal message.")
````

## Comparing strings

`==` tests equality. `<` and `>` compare **alphabetically** — more precisely, by character code.

That means **all uppercase letters sort before all lowercase letters**: `"Zebra" < "apple"` is
`True`, because `Z` (code 90) comes before `a` (code 97).

To compare fairly, normalise case first with `.lower()`.

````python
name = "Charlie"

if name < "M":
    print(name, "is in the first half of the alphabet")
else:
    print(name, "is in the second half")

print("Zebra" < "apple")
print("Zebra".lower() < "apple".lower())
````

## String **methods**

A method is a function that belongs to a value. You call it with a dot:

```python
text.upper()
```

Every string method **returns a new string** — remember, strings are immutable.

Use `dir(str)` to list them all, and `help(str.strip)` to read one.

````python
raw = "   Ada Lovelace\n"

print(repr(raw.upper()))
print(repr(raw.lower()))
print(repr(raw.strip()))         # removes whitespace from BOTH ends
print(repr(raw))                 # unchanged!
````

### Searching: `startswith`, `endswith`, `find`
`find` returns the **index** of the first match, or **`-1`** if there is none — not an error.

````python
line = "From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008"

print(line.startswith("From"))
print(line.endswith("2008"))
print(line.find("@"))          # index of the first @
print(line.find("zebra"))      # not found -> -1
print(line.find("a", 10))      # start searching at index 10
````

### Changing: `replace`
`replace(old, new)` returns a new string with **every** occurrence swapped.

````python
notice = "Class is cancelled on Monday. Monday only."

print(notice.replace("Monday", "Tuesday"))
print(notice.replace("Monday", "Tuesday", 1))   # only the first one
print(notice)                                    # original untouched
````

### Splitting: `split`

`split()` cuts a string into pieces at the whitespace and hands back a **list** of them.

We meet lists properly in **Week 9**. For today, all you need: the pieces come back numbered from
`0`, and you can grab one with `[ ]`, just like a string index.

````python
line = "Bianca Reyes 87 92 78"

parts = line.split()
print(parts)
print(parts[0])       # first name
print(parts[1])       # last name
print(len(parts))     # how many pieces
````

## Parsing a string

**Parsing** = finding the piece of text you actually want inside a larger piece.

The recipe:

1. Find a landmark with `find()`.
2. Slice relative to that landmark.
3. `strip()` off any stray whitespace.

Below: pull the **host name** out of an email log line — exactly the Chapter 6 example.

````python
line = "From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008"

at_pos = line.find("@")
space_pos = line.find(" ", at_pos)
host = line[at_pos + 1:space_pos]

print(at_pos, space_pos)
print(host)
````

## Live coding #1 — together

**Problem.** Write `initials(full_name)` that turns `"ada lovelace"` into `"A.L."`.

Rules: the name has exactly two words separated by a space; the result is the two capitalised
first letters, each followed by a period.

Think out loud before typing: where is the space? what do I slice?

````python
# Live coding #1
# initials("ada lovelace")  ->  "A.L."
# initials("Grace Hopper")  ->  "G.H."
````

````python
def initials(full_name):
    """Return the two initials of a two-word name, e.g. 'A.L.'."""
    space_pos = full_name.find(" ")
    first = full_name[0].upper()
    last = full_name[space_pos + 1].upper()
    return first + "." + last + "."


print(initials("ada lovelace"))
print(initials("Grace Hopper"))
````

## Making output look professional

So far our output has looked like this:

```
Total: 12.666666666666666
```

Nobody ships that. Today we fix it — permanently.

The tool: the **f-string**.

## f-strings

Put an `f` before the opening quote. Anything inside `{ }` is a **Python expression**, evaluated
and dropped into the text.

```python
name = "Sam"
print(f"Hello, {name}!")
```

Compare with what we have been doing:

```python
print("Hello, " + name + "!")        # concatenation — breaks on numbers
print("Hello,", name)                 # commas — you cannot control spacing
```

**From this week on, use f-strings.**

````python
name = "Sam"          # imagine the user typed this
score = 87
total = 100

print(f"Hello, {name}!")
print(f"{name} scored {score} out of {total}.")
print(f"That is {score / total * 100} percent.")
````

### Expressions inside the braces

Anything that produces a value works: arithmetic, method calls, function calls.

````python
name = "sam"
score = 87

print(f"{name.upper()} got {score} marks.")
print(f"Next mark up: {score + 1}")
print(f"Name length: {len(name)}")
print(f"Passed: {score >= 50}")
````

## Rounding to decimal places

Inside the braces, add `:` then a **format specification**.

`{value:.2f}` means *fixed-point, 2 digits after the decimal*.

This is how you print **money**. Always 2 decimals — `4.5` should display as `4.50`.

````python
subtotal = 19.99
tax_rate = 0.12
tax = subtotal * tax_rate
total = subtotal + tax

print(f"Tax:   {tax}")            # ugly
print(f"Tax:   ${tax:.2f}")       # money
print(f"Total: ${total:.2f}")
print(f"Rate:  {tax_rate:.1%}")   # percent, 1 decimal
````

`.2f` **rounds for display only** — the variable itself keeps its full precision. Never round your
stored value; round when you print.

## Field width and alignment

`{value:10}` reserves **10 characters** of space for the value.

| Spec | Meaning |
| :--- | :--- |
| `{x:10}` | width 10 — text left-aligns, numbers right-align |
| `{x:<10}` | force **left** align |
| `{x:>10}` | force **right** align |
| `{x:^10}` | **centre** |
| `{x:>8.2f}` | width 8, 2 decimals, right-aligned |
| `{x:,}` | thousands separators: `1,250,000` |

Width + alignment is how you build a **table** that lines up.

````python
label = "Mouse"

print(f"[{label:<12}]")
print(f"[{label:>12}]")
print(f"[{label:^12}]")
print(f"[{1250000:,}]")
print(f"[{3.14159:>9.3f}]")
````

### Putting it together: an aligned receipt

Names left, money right, decimals fixed. Notice how the header uses the *same* widths.

````python
print(f"{'Item':<14}{'Qty':>4}{'Amount':>10}")
print("-" * 28)
print(f"{'Wireless Mouse':<14}{2:>4}{2 * 24.99:>10.2f}")
print(f"{'USB-C Cable':<14}{3:>4}{3 * 9.50:>10.2f}")
print(f"{'Laptop Stand':<14}{1:>4}{1 * 129.00:>10.2f}")
print("-" * 28)
print(f"{'TOTAL':<14}{6:>4}{207.48:>10.2f}")
````

- `[    42]` — width 6, right-aligned by default for numbers anyway.

## `.format()` — the previous generation

You will see this in older code and in the textbook. Same specifiers **after the colon**, but the
values go in the argument list instead of inline.

```python
"Total: {:.2f}".format(total)
```

It still works, and it is useful when the template string is built separately from the values.
For everything you write, **prefer f-strings** — the value is right where you read it.

````python
name = "Sam"
total = 208.4832

print("{} owes ${:.2f}".format(name, total))
print("{0} owes ${1:.2f} — thanks, {0}!".format(name, total))
print(f"{name} owes ${total:.2f}")          # same thing, clearer
````

## The `%` operator — legacy, recognise but do not write

The oldest of the three. Chapter 6 calls it the *format operator*.

```python
"%d items cost %.2f dollars" % (3, 74.97)
```

`%d` = integer · `%s` = string · `%.2f` = float with 2 decimals. The values come after the `%`
in parentheses.

**Know it so you can read old code.** Write f-strings.

````python
count = 3
amount = 74.97

print("%d items cost %.2f dollars" % (count, amount))
print("Hello, %s!" % "Sam")
print(f"{count} items cost {amount:.2f} dollars")   # what we write today
````

## Common errors this week (1 of 2)

**1. `IndexError: string index out of range`**
You used index `len(s)`. The last index is `len(s) - 1`.

**2. `TypeError: 'str' object does not support item assignment`**
You wrote `word[0] = "J"`. Strings are immutable — build a new string:
`word = "J" + word[1:]`.

**3. The method ran but nothing changed.**
`text.upper()` on its own throws the result away. Methods **return** a new string:
`text = text.upper()`.

````python
name = "sam"
name.upper()          # result discarded!
print(name)

name = name.upper()   # this is what you meant
print(name)
````

## Common errors this week (2 of 2)

**4. `find()` returned `-1` and you sliced with it.**
No error is raised — you silently get the wrong text. Always check:

```python
pos = line.find("@")
if pos == -1:
    print("No @ in this line")
else:
    ...
```

**5. Forgetting the `f`.** `print("Total: {total:.2f}")` prints the braces literally.
No error, just wrong output. If you see `{` in your output, you dropped the `f`.

````python
total = 12.5

print("Total: {total:.2f}")     # forgot the f — prints the braces
print(f"Total: {total:.2f}")    # correct
````

## Summary

- A string is a **sequence**: `s[0]` is the first character, `s[-1]` the last,
  `len(s) - 1` the last valid index.
- **Slices** `s[a:b]` include `a`, exclude `b`, and never raise `IndexError`.
- Strings are **immutable** — every "change" returns a **new** string. Assign it or lose it.
- `in` tests substrings; `<` compares alphabetically and is **case-sensitive** — `.lower()` first.
- Methods to know: `upper`, `lower`, `strip`, `startswith`, `find` (returns `-1`!), `replace`,
  `split`.
- **f-strings** are the way we format from now on: `{value:.2f}` for money,
  `{text:<14}` and `{n:>8}` for aligned tables.
