## A list is a sequence

A **list** is an ordered collection of values. Anything you could keep in a
pile — songs on a playlist, quiz scores, temperatures for the week — belongs in
a list instead of in ten separate variables.

- Written with **square brackets**, items separated by commas.
- Items are **indexed from 0**, exactly like string characters.
- The items are called **elements**.

`scores = [88, 74, 95, 61]`

````python
scores = [88, 74, 95, 61]
songs = ["Levitating", "Bad Habit", "Anti-Hero"]

print(scores)
print(songs)
print(scores[0])      # first element
print(songs[2])       # third element
print(songs[-1])      # last element
print(len(scores))    # how many elements
````

### A list can hold anything — even other lists

An element can be a number, a string, a boolean, or another list. A list inside
a list is called a **nested list**, and it needs two indexes.

An **empty list** is written `[]` — very useful as a starting point you then
build up.

````python
mixed = ["Sam", 19, True, 3.5]
print(mixed[1], mixed[3])

gradebook = [["Sam", 88], ["Ana", 74], ["Kim", 95]]
print(gradebook[1])        # the whole inner list
print(gradebook[1][0])     # Ana
print(gradebook[1][1])     # 74

empty = []
print(empty, len(empty))
````

## Lists are **mutable** — strings are not

This is the headline difference from Week 7.

| | String (Week 7) | List (this week) |
| :--- | :--- | :--- |
| Index with `[i]` | yes | yes |
| Slice with `[a:b]` | yes | yes |
| `len()`, `in`, `for` | yes | yes |
| **Assign to `x[i]`** | **TypeError** | **works** |

A string method like `upper()` **returns a new string**. A list method like
`append()` **changes the list you already have**.

````python
name = "sam"
scores = [88, 74, 95]

scores[1] = 100          # allowed: lists are mutable
print(scores)

print(name.upper())      # returns a NEW string
print(name)              # the original is unchanged
````

````python
# Strings refuse in-place assignment.
name = "sam"
name[0] = "S"
````

## Traversing a list

Two shapes, exactly as with strings in Week 7:

1. **`for item in my_list:`** — when you only need each *value*.
2. **`for i in range(len(my_list)):`** — when you need the *position*, usually
   because you want to **change** the element.

Shape 1 is the one you want most of the time. Reach for shape 2 only when you
need `i`.

````python
temperatures = [12, 15, 9, 21]

total = 0
for reading in temperatures:
    total = total + reading
print("Total:", total)

# Need the index because we are modifying the list in place.
for i in range(len(temperatures)):
    temperatures[i] = temperatures[i] + 1
print(temperatures)
````

### `in` works on lists too

`in` asks *is this value an element of the list?* It compares whole elements —
not parts of them.

````python
playlist = ["Levitating", "Bad Habit", "Anti-Hero"]

print("Anti-Hero" in playlist)
print("Anti" in playlist)          # False: not a whole element
print("Anti" in "Anti-Hero")       # True: substring of a STRING

if "Bad Habit" not in playlist:
    print("adding it")
else:
    print("already on the playlist")
````

## List operations: `+` and `*`

- `+` **concatenates**: builds a **new** list containing the elements of both.
- `*` **repeats**: `[0] * 5` builds a new list of five zeros.

Both build new lists — neither changes the originals. `[0] * n` is the standard
way to create a list of a known size filled with a starting value.

````python
monday = ["eggs", "toast"]
tuesday = ["rice", "beans"]

groceries = monday + tuesday
print(groceries)
print(monday)               # unchanged

counters = [0] * 5
print(counters)
print(["ha"] * 3)
````

## List slices

`my_list[start:end]` gives a **new list** with the elements from `start` up to
**but not including** `end` — identical rules to string slicing in Week 7.

- Leave off `start` → from the beginning.
- Leave off `end` → to the end.
- `my_list[:]` → a full slice, i.e. **a copy of the whole list**.

````python
letters = ["a", "b", "c", "d", "e", "f"]

print(letters[1:4])     # b c d  -- 4 is NOT included
print(letters[:3])      # first three
print(letters[3:])      # rest
print(letters[-2:])     # last two
print(letters[:])       # a copy of everything
````

### Slices on the *left* of `=`

Because lists are mutable, a slice can be the **target** of an assignment: it
replaces that whole stretch of the list. Strings can never do this.

````python
letters = ["a", "b", "c", "d", "e", "f"]
letters[1:3] = ["X", "Y"]
print(letters)

# The replacement need not be the same length.
letters[1:3] = ["Q"]
print(letters)
````

## List methods that **add** elements

A method is called on the list with a dot: `my_list.append(x)`.

| Method | What it does |
| :--- | :--- |
| `append(x)` | adds **one** element `x` to the end |
| `extend(other)` | adds **every element of** `other` to the end |
| `insert(i, x)` | puts `x` at position `i`, shifting the rest right |

All three **modify the list in place and return `None`**. Never write
`scores = scores.append(90)`.

````python
scores = [88, 74]

scores.append(95)
print(scores)

scores.extend([61, 70])       # each element added separately
print(scores)

scores.append([50, 55])       # ONE element, which is a list
print(scores)

scores.insert(0, 100)
print(scores)
````

## Removing elements

| Way | Use when |
| :--- | :--- |
| `pop(i)` | you want the element **back**; `pop()` with no argument takes the last |
| `remove(x)` | you know the **value**, not the position; removes the **first** match |
| `del my_list[i]` | you know the position and don't need the value |
| `del my_list[a:b]` | you want to delete a whole **slice** |

`pop` is the only one that hands the value back to you.

````python
queue = ["Sam", "Ana", "Kim", "Lee"]

first = queue.pop(0)
print("served:", first, "->", queue)

last = queue.pop()
print("served:", last, "->", queue)

queue.remove("Ana")           # by value
print(queue)

snacks = ["chips", "pop", "candy", "gum"]
del snacks[1]
print(snacks)
del snacks[0:2]
print(snacks)
````

## Sorting

- `my_list.sort()` sorts the list **in place** and returns `None`.
- `sorted(my_list)` returns a **new sorted list** and leaves the original alone.
- `my_list.reverse()` reverses in place.

Use `sort()` when you no longer need the original order; use `sorted()` when you
do.

````python
scores = [88, 74, 95, 61]

print(sorted(scores))        # new list
print(scores)                # original untouched

scores.sort()
print(scores)                # now the original IS sorted

scores.sort(reverse=True)
print(scores)

names = ["kim", "Ana", "Sam"]
names.sort()
print(names)                 # capitals sort before lowercase
````

## Lists and built-in functions

Week 5 taught you to accumulate a total, a maximum, and a minimum with a loop.
For lists, Python has already written those loops:

| Function | Result |
| :--- | :--- |
| `len(xs)` | how many elements |
| `sum(xs)` | total (numbers only) |
| `max(xs)` / `min(xs)` | largest / smallest |

Knowing the loop *and* the built-in matters: the midterm may ask for either.

````python
temperatures = [12, 15, 9, 21, 18]

print("count:  ", len(temperatures))
print("total:  ", sum(temperatures))
print("highest:", max(temperatures))
print("lowest: ", min(temperatures))

average = sum(temperatures) / len(temperatures)
print("average:", round(average, 2))
````

### Guard against the empty list

`sum([])` is `0`, but `max([])` **raises an error**, and dividing by
`len([])` is a `ZeroDivisionError`. Real data has empty cases — check first.

````python
def average(numbers):
    """Return the mean of a list of numbers, or 0.0 if it is empty."""
    if len(numbers) == 0:
        return 0.0
    return sum(numbers) / len(numbers)


print(average([12, 15, 9]))
print(average([]))
````

## Lists and strings: `split()`

`split()` turns a **string** into a **list of words**.

- With no argument it splits on any run of whitespace and drops the blanks.
- With an argument — `line.split(",")` — it splits on that exact **delimiter**.

This is how a line of a file becomes usable data. It closes the loop from
Week 7 (strings) and Week 8 (files).

````python
message = "the quick brown fox"
words = message.split()
print(words)
print(len(words), "words")

row = "Sam,88,COMP1110"
fields = row.split(",")
print(fields)
print("student:", fields[0], "score:", int(fields[1]))
````

### `join()` — the opposite direction

`join` is called on the **separator string**, and takes the **list**:
`", ".join(words)`. It only works on a list of strings.

````python
words = ["the", "quick", "brown", "fox"]

print(" ".join(words))
print("-".join(words))
print("".join(words))

numbers = [88, 74, 95]
as_text = []
for n in numbers:
    as_text.append(str(n))       # join needs strings
print(", ".join(as_text))
````

## Parsing lines — Weeks 7, 8 and 9 together

The standard pattern for a data file:

1. `open` the file and loop over its lines *(Week 8)*.
2. `rstrip()` the trailing newline *(Week 7)*.
3. `split()` the line into a list of fields *(this week)*.
4. Convert the fields you need with `int()` or `float()` and accumulate.

We first **write** a small file so the notebook is self-contained.

````python
lines = "Sam,88\nAna,74\nKim,95\nLee,61\n"
handle = open("week09_grades.txt", "w")
handle.write(lines)
handle.close()

handle = open("week09_grades.txt")
for line in handle:
    print(repr(line))
handle.close()
````

````python
# Now parse it into two parallel lists.
names = []
scores = []

handle = open("week09_grades.txt")
for line in handle:
    line = line.rstrip()
    if line == "":
        continue
    fields = line.split(",")
    names.append(fields[0])
    scores.append(int(fields[1]))
handle.close()

print(names)
print(scores)
print("class average:", round(sum(scores) / len(scores), 1))
print("top score:", max(scores))
````

## Objects and values: `is` vs `==`

Two different questions:

- `a == b` — do they hold the **same value**?
- `a is b` — are they the **same object** in memory?

For strings, Python often reuses one object, so `is` may surprise you. For lists,
**two lists written separately are always two different objects**.

`id(x)` shows the identity Python uses. Use `==` in your programs; `is` is for
reasoning about aliasing (and for `is None`).

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)      # same value
print(a is b)      # different objects
print(id(a) == id(b))

c = a              # same object, two names
print(c == a, c is a)
````

## Aliasing — the heart of this week

When you write `b = a` and `a` refers to a list, **no copy is made**. Both names
point at the *same* list object. `b` is an **alias** for `a`.

```
a ──▶ [ 88, 74, 95 ] ◀── b
```

Change the list through either name and **both** names see the change, because
there is only one list.

This never bit you with strings or numbers — they are **immutable**, so there is
nothing to change.

````python
a = [88, 74, 95]
b = a                 # alias, NOT a copy

b.append(100)
print("a:", a)
print("b:", b)
print("same object?", a is b)
````

### Why numbers and strings feel different

````python
x = 5
y = x
y = y + 1             # rebinds y to a NEW int; x is untouched
print(x, y)

s = "sam"
t = s
t = t.upper()         # rebinds t to a NEW string
print(s, t)

# Mutation vs rebinding is the whole distinction:
p = [1, 2]
q = p
q = q + [3]           # rebinding: q now points somewhere else
print(p, q, p is q)
````

### Making a real copy

Three equivalent ways to get an **independent** list:

- `b = a[:]`  (a full slice)
- `b = list(a)`
- `b = a.copy()`

Any of these gives you a new object, so mutating `b` leaves `a` alone.

````python
a = [88, 74, 95]

b = a[:]
c = list(a)
d = a.copy()

b.append(100)
print("a:", a)
print("b:", b)
print(a is b, a is c, a is d)
print(a == c)
````

## Predict the output #3 — a function that mutates

```python
def add_bonus(marks):
    for i in range(len(marks)):
        marks[i] = marks[i] + 5
    return marks


original = [70, 80]
adjusted = add_bonus(original)

print(original)
print(adjusted)
print(original is adjusted)
```

### Answer

```
[75, 85]
[75, 85]
True
```

The parameter `marks` is an **alias** for `original`. The loop mutates the one
and only list, so the caller's list changed too — even though the caller never
asked for that.

A function that changes its argument is said to have a **side effect**. It is
legal and sometimes what you want, but it must be **documented**, and a function
should either mutate *or* return a new list — not quietly do both.

### The pure alternative

Build and return a **new** list; leave the argument untouched. Compare the two
docstrings — this is exactly the documentation habit from Week 8.

````python
def with_bonus(marks):
    """Return a NEW list with 5 added to every mark. Does not modify `marks`."""
    result = []
    for mark in marks:
        result.append(mark + 5)
    return result


original = [70, 80]
adjusted = with_bonus(original)
print(original)          # safe
print(adjusted)
````

## The mutable default argument trap

```python
def add_song(title, playlist=[]):
    ...
```

Looks harmless. It is not.

The default `[]` is created **once**, when the `def` line runs — **not** on each
call. Every call that omits the argument shares that same list, so it fills up
across calls.

````python
def add_song(title, playlist=[]):
    playlist.append(title)
    return playlist


print(add_song("Levitating"))
print(add_song("Anti-Hero"))       # expected one song, got two
print(add_song("Bad Habit"))
````

### The fix: default to `None`

Use `None` as the default and create a fresh list inside the function. This is
the standard Python idiom — expect to see it in your project code.

````python
def add_song(title, playlist=None):
    """Return a playlist with `title` added; start a new one if none given."""
    if playlist is None:
        playlist = []
    playlist.append(title)
    return playlist


print(add_song("Levitating"))
print(add_song("Anti-Hero"))
print(add_song("Bad Habit", ["Older song"]))
````

## Live coding 1 — word statistics

Together, from a blank cell.

Write `word_report(text)` that:

1. splits `text` into a list of words,
2. prints how many words there are,
3. prints the **longest** word,
4. prints the words joined back together with `" | "`.

Test it on: `"the quick brown fox jumps over the lazy dog"`.

````python
# Live coding 1 — word_report(text)
# 1. split into words   2. count   3. longest word   4. join with " | "
````

````python
def word_report(text):
    """Print the word count, the longest word, and the words rejoined."""
    words = text.split()
    print("word count:", len(words))

    longest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word
    print("longest word:", longest)

    print(" | ".join(words))


word_report("the quick brown fox jumps over the lazy dog")
````

## Live coding 2 — top scores from a file

Write `top_scores(filename, cutoff)` that reads `week09_grades.txt`
(`name,score` per line) and **returns a list of the names** whose score is at
least `cutoff`.

Then print the returned list, and the class average.

Watch for: stripping the newline, `int()` conversion, and *not* mutating
anything the caller owns.

````python
# Live coding 2 — top_scores(filename, cutoff) -> list of names
````

````python
def top_scores(filename, cutoff):
    """Return a list of names from `filename` whose score is >= cutoff."""
    winners = []
    handle = open(filename)
    for line in handle:
        line = line.rstrip()
        if line == "":
            continue
        fields = line.split(",")
        if int(fields[1]) >= cutoff:
            winners.append(fields[0])
    handle.close()
    return winners


print(top_scores("week09_grades.txt", 80))
print(top_scores("week09_grades.txt", 100))
````

## Common errors this week

### 1. Assigning the result of a mutating method

```python
scores = scores.sort()      # scores is now None
```

`sort`, `append`, `extend`, `insert`, `remove`, and `reverse` all return `None`.
Call them as statements. The symptom is
`TypeError: 'NoneType' object is not iterable` (or `not subscriptable`) on a
*later* line.

````python
scores = [88, 74, 95]
scores = scores.sort()
print(scores[0])
````

### 2. Index out of range

Valid indexes run `0` to `len(xs) - 1`. `xs[len(xs)]` is always one too far —
the classic off-by-one.

**`IndexError: list index out of range`**

````python
letters = ["a", "b", "c"]
print(letters[len(letters)])
````

### 3. `remove` on a value that is not there

`remove(x)` needs `x` to actually be an element. Check with `in` first.

**`ValueError: list.remove(x): x not in list`**

````python
playlist = ["Levitating", "Anti-Hero"]
playlist.remove("Bad Habit")
````

### 4. `join` on a list of numbers

`join` glues **strings**. Convert first with `str()` in a loop.

**`TypeError: sequence item 0: expected str instance, int found`**

````python
scores = [88, 74, 95]
print(", ".join(scores))
````

## Summary

- A **list** is an ordered, indexed sequence — like a string, but **mutable**.
  `xs[i] = v` works; `s[i] = c` raises `TypeError`.
- Add with `append` / `extend` / `insert`; remove with `pop` / `remove` / `del`;
  order with `sort()` (in place) or `sorted()` (new list). Mutating methods
  return **`None`**.
- `split()` turns a line into a list of fields; `join()` turns a list of strings
  back into text. That is how file data becomes usable.
- `b = a` creates an **alias**, not a copy. Use `a[:]`, `list(a)`, or `a.copy()`
  for a real copy, and `is` to tell objects apart from equal values.
- A function **can** modify a list argument. Say so in the docstring, or return a
  new list instead. Never default a parameter to `[]`.
