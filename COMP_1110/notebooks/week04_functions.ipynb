## Part 1 — Calling functions

You have been calling functions since Week 1: `print(...)`, `input(...)`, `int(...)`.

A **function call** is a **name**, then parentheses, then **arguments**:

```
      type( 42 )
       │     │
       │     └── the argument: the value you hand in
       └──────── the function's name
```

A call **is an expression** — it evaluates to a value you can store or use further.

## Built-in functions

Always available, no import needed:

`print` · `len` · `max` · `min` · `abs` · `round` · `type` · `int` · `float` · `str`

You are **not** expected to memorise every built-in. You are expected to
recognise a call when you see one.

````python
song_title = "Bohemian Rhapsody"

print(len(song_title))
print(max(72, 88, 65))
print(min(72, 88, 65))
print(abs(-4.5))
print(round(3.14159, 2))
print(type(song_title))
````

### A call does not change its argument

`round` **gives back** a new value; it does not edit `price` in place.

````python
price = 19.987
round(price, 2)          # the answer is computed, then thrown away
print(price)

price = round(price, 2)  # this is how you keep it
print(price)
````

## Type conversion functions

`input()` always gives a **string**, and arithmetic on strings is not arithmetic.

- `int(x)` → whole number · `float(x)` → decimal number · `str(x)` → text

`int()` on a float **truncates toward zero** — it does **not** round.

````python
reply = "17"          # imagine the user typed this at an input() prompt
age = int(reply)
print(age + 1)

print(int(3.99))      # truncates, does not round
print(int(-3.99))
print(float("3.5") + 1)
print("Age: " + str(age))
````

### When the text has a decimal point

`int("3.9")` is a `ValueError`. Convert via `float` first.

````python
print(int(float("3.9")))
print(int(" 42 "))        # surrounding spaces are fine
````

## The `math` module

A **module** is a file of ready-made functions. Import once, then use
**dot notation**: `module.function(...)`.

```python
import math
```

`math.pi` has **no parentheses** — it is a value, not a function.

````python
import math

print(math.sqrt(144))
print(math.floor(7.8))    # down
print(math.ceil(7.2))     # up
print(round(math.pi, 4))

# Distance from the origin to the point (3, 4)
print(math.sqrt(3 ** 2 + 4 ** 2))
````

## Random numbers

Computers are deterministic, so `random` gives **pseudorandom** numbers: a fixed
sequence that merely *looks* unpredictable.

`random.random()` → float, `0.0 <= x < 1.0` · `random.randint(1, 6)` → whole
number 1–6 **inclusive** · `random.uniform(a, b)` → float between `a` and `b`

`random.seed(n)` fixes the starting point so results **repeat exactly** — which
is why this notebook's output is stable.

````python
import random

random.seed(1110)        # same seed => same numbers every run

print(random.randint(1, 6))
print(random.randint(1, 6))
print(round(random.random(), 3))
print(round(random.uniform(2.0, 5.0), 3))
````

### Watch the boundaries

`randint(1, 6)` **includes** 6. `random()` **never** returns 1.0.
Off-by-one at the boundary is a classic bug — check the docs, do not guess.

## Part 2 — Writing your own functions

```python
def function_name(parameters):
    # body: indented 4 spaces
    ...
```

- `def` is the keyword · the **header** ends with a **colon**
- the **body** is **indented** 4 spaces (PEP 8)
- the name follows variable-name rules: `snake_case`

````python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")


print_lyrics()
print_lyrics()
````

### Two separate events

- The `def` statement **creates** the function. Nothing inside runs yet.
- The call `print_lyrics()` **runs** the body.

Defining a function you never call produces **no output at all**.

## Definitions and uses — order matters

Python reads your file **top to bottom**. A function must be **defined before it
is called**:

```python
greet_student()      # NameError — nothing by that name exists yet

def greet_student():
    print("Welcome to COMP 1110")
```

Rule of thumb: **all your `def`s at the top**, the code that calls them at the bottom.

````python
def print_header():
    print("=" * 30)
    print("  TRU Grade Report")


def print_footer():
    print("=" * 30)


def print_report_shell():     # one function may call another
    print_header()
    print("  (marks go here)")
    print_footer()


print_report_shell()
````

## Flow of execution

When a call happens, Python:

1. **pauses** where it is, 2. **jumps** into the body, 3. runs the body top to
bottom, 4. **returns to exactly where it left off** and carries on.

You should be able to trace this by hand — the midterm asks you to.

````python
def middle_step():
    print("2. inside middle_step")


def outer_step():
    print("1. inside outer_step, about to call middle_step")
    middle_step()
    print("3. back in outer_step")


print("0. before the call")
outer_step()
print("4. after the call")
````

## Parameters and arguments

- A **parameter** is the name in the **definition**.
- An **argument** is the value supplied at the **call**.

At call time the argument is **assigned** to the parameter. A parameter is just
a variable that lives only inside the function.

Same idea, two words — use the right one in your answers.

````python
def greet(name, course):
    print("Hello " + name + ", welcome to " + course + ".")


greet("Amara", "COMP 1110")

student = "Priya"
greet(student, "COMP 1110")   # arguments can be variables, not just literals
````

### Arguments are matched by **position**

Swap them and Python will not complain — it will simply do the wrong thing.
A **logic error**: no error message will save you.

````python
def greet(name, course):
    print("Hello " + name + ", welcome to " + course + ".")


greet("COMP 1110", "Diego")   # runs perfectly. Also nonsense.
````

## Part 3 — `return` vs `print`
### The single biggest misconception of Week 4

- **`print`** puts characters on the **screen**, for a **human** to read.
- **`return`** hands a **value** back to the **code that made the call**, for the
  **program** to use.

> A function that prints has *told* you the answer.
> A function that returns has *given* you the answer.

````python
def add_tax_printing(price):
    print(round(price * 1.12, 2))    # shows it


def add_tax_returning(price):
    return price * 1.12              # hands it back


add_tax_printing(100)                # something appears on screen
with_tax = add_tax_returning(100)    # nothing appears; the value is in `with_tax`
print(round(with_tax, 2))
````

### The test that settles it

Can you **use the answer for something else**? Only the returning version can be
fed into more arithmetic.

````python
def add_tax_returning(price):
    return price * 1.12


print(round(add_tax_returning(100) + add_tax_returning(50), 2))
````

## Predict the output #2 — `return` vs `print`

```python
def double_print(n):
    print(n * 2)


def double_return(n):
    return n * 2


a = double_print(5)
b = double_return(5)

print(a)
print(b)
print(a == b)
```

Four lines of output. Write down all four.

### Answer

```
10
None
10
False
```

1. `double_print(5)` **prints** `10` as it runs. It has no `return`, so it hands
   back **`None`** — that is what `a` becomes.
2. `double_return(5)` prints nothing; `b` becomes `10`.
3. `print(a)` → `None` · `print(b)` → `10` · `None == 10` → `False`

**A function with no `return` statement returns `None`.** Always.

## Fruitful functions and void functions

- **Fruitful:** has a `return`, produces a value — `math.sqrt`, `len`, `int`.
- **Void:** no `return`, produces `None` — `print`, and every function of yours
  that only prints.

`None` is a real value of its own type, `NoneType`. It means *"no value here"* —
it is **not** the string `"None"`, and **not** zero.

````python
def show_grade(mark):
    print("Your mark is " + str(mark))


result = show_grade(88)
print(result)
print(type(result))
print(result is None)
````

### Why this bites people

```python
total = show_grade(88)
print(total + 10)
# TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'
```

You cannot do arithmetic on `None`. That message almost always means a function
**printed** where it should have **returned**.

## `return` ends the function immediately

Nothing after a `return` on the same path ever runs. That makes `return` inside
an `if` very natural: each branch hands back its own answer and stops there.

````python
def letter_grade(mark):
    if mark >= 80:
        return "A"
    elif mark >= 65:
        return "B"
    elif mark >= 50:
        return "C"
    return "F"          # reached only if every condition above was False


print(letter_grade(91))
print(letter_grade(66))
print(letter_grade(20))
````

## Part 4 — Why functions? *[CLO 2]*

Writing `def` is easy. Knowing **what deserves to be a function** is the skill.

1. **Naming** — six confusing lines become one line that says what it means.
2. **Removing repetition** — write it once, fix bugs in one place.
3. **Decomposition** — split a hard problem into small ones you can test
   **separately**.
4. **Reuse** — today's function is next week's building block.

> If you copy, paste, and edit a number, you wanted a function with a parameter.

## Decomposition, worked

**Problem:** a café bill. Given a price and a tip percent, print a receipt showing
subtotal, tip, tax, and total.

Do **not** start typing. Name the pieces first:

| Sub-problem | Function | Kind |
| :--- | :--- | :--- |
| How much tip? | `tip_amount(subtotal, percent)` | fruitful |
| How much tax? | `tax_amount(subtotal)` | fruitful |
| Show the bill | `print_receipt(subtotal, percent)` | void |

````python
TAX_RATE = 0.12


def tip_amount(subtotal, percent):
    return subtotal * percent / 100


def tax_amount(subtotal):
    return subtotal * TAX_RATE
````

````python
def print_receipt(subtotal, percent):
    tip = tip_amount(subtotal, percent)
    tax = tax_amount(subtotal)
    print("Subtotal: " + str(round(subtotal, 2)))
    print("Tip:      " + str(round(tip, 2)))
    print("Tax:      " + str(round(tax, 2)))
    print("Total:    " + str(round(subtotal + tip + tax, 2)))


print_receipt(48.50, 18)
````

### Notice the shape

- The **fruitful** functions know nothing about printing.
- The **void** function knows nothing about how tip or tax are computed.

Change the tax rate → edit **one line**. Change the layout → touch **one
function**. That is what "modular" means, and it is exactly what **CLO 2** asks.

## Part 5 — Scope: where a variable lives

A variable created **inside** a function exists **only inside that function**, and
is **discarded** when the function returns.

- **Local:** created inside a function — including every **parameter**.
- **Global:** created at the top level of the file.

Each call gets a **fresh** set of locals. Two calls never share them.

````python
def compute_average(a, b):
    total = a + b          # `total` is LOCAL to compute_average
    return total / 2


print(compute_average(80, 90))
print(compute_average(60, 70))   # a fresh `total` each time
````

### Asking for a local from outside

`total` was thrown away the moment `compute_average` returned. The next cell
raises **`NameError`** — deliberately.

````python
def compute_average(a, b):
    total = a + b
    return total / 2


compute_average(80, 90)
print(total)      # NameError: name 'total' is not defined
````

## Reading a global vs assigning to one

A function may **read** a global it did not create.

But the moment a function **assigns** to a name anywhere in its body, Python
treats that name as **local for the whole function** — including lines *before*
the assignment. Reading it first gives **`UnboundLocalError`**.

````python
MAX_MARK = 100          # global


def percentage(score):
    return score / MAX_MARK * 100   # reading a global: fine


print(percentage(45))
````

````python
counter = 0


def bump():
    print(counter)          # UnboundLocalError: `counter` is local here...
    counter = counter + 1   # ...because this line assigns to it


bump()
````

### How to fix it — and what not to do

The `global` keyword exists, but reaching for it is a design smell in this
course. **Pass the value in and return the new one:**

```python
def bump(counter):
    return counter + 1

counter = bump(counter)
```

Data flows **in through parameters**, **out through `return`**. Keep it that way
and scope will rarely surprise you.

### Same name, different variable

A local name **shadows** a global one. Assigning to the local does not touch the
global.

````python
message = "global message"


def show():
    message = "local message"     # a brand-new local variable
    print(message)


show()
print(message)      # the global is untouched
````

## Live coding #1 — together

Write a **fruitful** `celsius_to_fahrenheit(celsius)` returning `C * 9 / 5 + 32`.

Then a **void** `report_temperature(celsius)` that prints:

```
21 C is 69.8 F (mild)
```

`cold` below 10, `mild` from 10 to 24, `hot` at 25 or above.

**Constraints:** the conversion must not print; the report must not do the
arithmetic itself.

````python
# Live coding #1 — we write this together in class.
#
# def celsius_to_fahrenheit(celsius):
#     ...
#
# def report_temperature(celsius):
#     ...
#
# report_temperature(21)
````

````python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32


def describe_temperature(celsius):
    if celsius < 10:
        return "cold"
    elif celsius < 25:
        return "mild"
    return "hot"


def report_temperature(celsius):
    fahrenheit = celsius_to_fahrenheit(celsius)
    label = describe_temperature(celsius)
    print(str(celsius) + " C is " + str(round(fahrenheit, 1))
          + " F (" + label + ")")


report_temperature(21)
report_temperature(3)
report_temperature(31)
````

## Live coding #2 — together

A dice game. Write:

- `roll_die()` — no arguments, **returns** a random whole number 1–6
- `roll_two()` — **returns** the total of two dice (must call `roll_die`)
- `play_round()` — **prints** the total and whether it is a win (7 or 11),
  a loss (2, 3, 12), or "roll again"

**Ask yourselves:** which are fruitful, and why must `roll_die` return rather
than print?

````python
# Live coding #2 — we write this together in class.
#
# import random
#
# def roll_die():
#     ...
#
# def roll_two():
#     ...
#
# def play_round():
#     ...
````

````python
import random

random.seed(7)


def roll_die():
    return random.randint(1, 6)


def roll_two():
    return roll_die() + roll_die()


def play_round():
    total = roll_two()
    if total == 7 or total == 11:
        print("Rolled " + str(total) + " — you win!")
    elif total == 2 or total == 3 or total == 12:
        print("Rolled " + str(total) + " — you lose.")
    else:
        print("Rolled " + str(total) + " — roll again.")


play_round()
play_round()
play_round()
````

## Common errors this week

### 1. Forgot the parentheses on the call

```python
print_lyrics        # does nothing visible
print_lyrics()      # calls it
```

Without `()` you refer to the function **object** instead of calling it.
No error, no output — the most confusing failure of the week.

### 2. `TypeError: greet() missing 1 required positional argument: 'name'`

```python
def greet(name):
    print("Hi " + name)

greet()             # TypeError
```

**Meaning:** the definition asks for one argument; you gave none. Counts must
match exactly. The mirror image is
`takes 1 positional argument but 2 were given`.

### 3. `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`

```python
def total_price(a, b):
    print(a + b)        # should be `return`

grand = total_price(10, 5) + 2   # TypeError
```

**Meaning:** you used the result of a function that returns nothing.
**Fix:** change `print` to `return`, and print at the *call site* instead.

### 4. `IndentationError: expected an indented block`

```python
def greet(name):
print("Hi " + name)     # not indented
```

**Meaning:** the line after a `def ...:` header must be indented 4 spaces, and
every statement in the body keeps that indentation.

And: `NameError: name 'greet' is not defined` — you called it **above** its
`def`, or misspelled it. Python names are **case-sensitive**.

## Debugging functions *[CLO 4]*

When a function misbehaves, check in this order:

1. **Is it even being called?** `print("entered")` as the first line of the body.
2. **What did it receive?** Print each parameter at the top of the body.
3. **What is it handing back?** `print(type(result))` at the call site — if that
   says `NoneType`, you printed where you meant to return.
4. **Test one function at a time.** That is the payoff of decomposition: call
   `tip_amount(100, 15)` alone and check the number by hand.

Read tracebacks **from the bottom up**.

## Summary

- A **call** is a name plus parentheses plus arguments, and it evaluates to a
  value. `import math` / `import random` add more callable functions.
- `def` **defines**; a call **runs**. Define before you call; execution jumps in,
  runs the body, and returns to where it left off.
- **Parameters** are names in the definition, **arguments** are values at the
  call, matched by **position**.
- **`return` hands a value back; `print` shows characters.** No `return` means
  **`None`** — the source of most `NoneType` errors.
- Locals **vanish** when a function returns; asking for one outside is a
  **`NameError`**.
- Decompose: small fruitful functions that compute, one void function that
  displays. *[CLO 2]*
