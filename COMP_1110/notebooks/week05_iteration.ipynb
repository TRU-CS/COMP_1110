## Updating variables

Everything about loops rests on one idea from Week 2:

```python
x = x + 1
```

This is **not** an equation — it is an instruction. Read it right-to-left:

> "Take the current value of `x`, add 1, and store the result back in `x`."

Names for these:

- `x = x + 1` — an **increment**
- `x = x - 1` — a **decrement**
- `total = total + price` — an **accumulator**

A variable you update this way must be **initialised first**.

````python
# An accumulator: adding up a shopping basket, one item at a time.
total = 0.0                 # initialise BEFORE updating

total = total + 4.99        # milk
print("after milk: ", total)

total = total + 12.50       # cheese
print("after cheese:", total)

total = total + 2.25        # bread
print("after bread: ", total)

print("basket total:", round(total, 2))
````

Notice `total` is *reassigned* three times. There is only ever one `total`;
each line replaces its value with a new one computed from the old one.

A loop does exactly this — but the update line is written **once** and executed
many times.

## The `while` statement

An **indefinite loop**: we do not know in advance how many times it will run.
It runs *while* a condition stays `True`.

```python
while <boolean expression>:
    <indented body>
```

Every correct `while` loop has **three** parts:

1. **Initialise** the loop variable *before* the loop.
2. **Test** it in the condition.
3. **Update** it *inside* the body, so the condition can eventually become
   `False`.

Miss part 3 and the loop never ends.

````python
# A countdown. Initialise, test, update.
seconds = 3                     # 1. initialise

while seconds > 0:              # 2. test
    print("T minus", seconds)
    seconds = seconds - 1       # 3. update

print("Liftoff!")
````

### Hand-tracing that loop

The midterm is closed-book with **no running Python**. So we trace by hand:
one row per pass, one column per variable.

| Pass | `seconds` at test | Test `seconds > 0` | Printed | `seconds` after update |
| :--- | :--- | :--- | :--- | :--- |
| 1 | 3 | `True` | `T minus 3` | 2 |
| 2 | 2 | `True` | `T minus 2` | 1 |
| 3 | 1 | `True` | `T minus 1` | 0 |
| 4 | 0 | `False` | — | loop exits |

Then `print("Liftoff!")` runs once. **Draw this table on paper.** It is the
single most reliable way to answer a trace question.

Two habits worth building now:

- The condition is checked **before** every pass, including the first.
- The loop body may run **zero** times. Set `seconds = 0` above and nothing is
  printed except `Liftoff!`.

## Infinite loops — and why we will not run one

An **infinite loop** is a loop whose condition never becomes `False`.

```python
# DO NOT RUN THIS.
n = 10
while n > 0:
    print("Still going...")   # n is never updated!
```

`n` stays 10 forever. The program prints until you kill it. In Jupyter the cell
shows `[*]` and never finishes; the output area grows until the browser tab
freezes.

**How to escape:** press the ⏹ *Interrupt* button (or *Kernel → Interrupt*).
In a terminal, press **Ctrl-C**.

The three ways students actually create one:

1. **No update.** The variable in the condition is never changed in the body.
2. **Update in the wrong direction.** `while n > 0:` with `n = n + 1`.
3. **Update outside the loop.** The update line is not indented into the body,
   so it runs only once, after the loop would have finished.

Before you run a `while` loop, ask out loud: *"what makes this stop?"* If you
cannot answer in one sentence, do not run it.

## `break` — leave the loop now

`break` immediately exits the **innermost** loop. Execution continues at the
first statement after the loop body.

This gives us the **loop-and-a-half**: an intentionally infinite `while True:`
whose only exit is a `break` in the middle. It is the standard shape for
"keep asking until the user says stop".

```python
while True:
    ...get something...
    if <it is time to stop>:
        break
    ...use it...
```

A `while True:` with no `break` is an infinite loop. The `break` *is* the
condition.

````python
# Setup: canned answers so input() never blocks.
def use_canned(*values):
    """Replace input() with one that returns each given value in turn."""
    global input
    remaining = iter(values)

    def canned_input(prompt=""):
        answer = next(remaining)
        print(prompt + answer)
        return answer

    input = canned_input


use_canned("hello", "see you tuesday", "done")
````

````python
# Echo text messages until the user types 'done'.
while True:
    message = input("Message (or 'done'): ")
    if message == "done":
        break
    print("   sent:", message)

print("Chat closed.")
````

## `continue` — skip the rest of *this* pass

`continue` abandons the current pass and jumps straight back to the top of the
loop, re-testing the condition.

- `break` — leave the loop entirely.
- `continue` — skip to the next iteration.

It is how you say *"this item is not interesting, move on"* without wrapping
the whole body in an `else`.

> **Warning:** in a `while` loop, `continue` skips the rest of the body —
> including your update line, if it is at the bottom. That is a classic way to
> write an accidental infinite loop.

````python
use_canned("hi there", "", "", "running late", "done")
````

````python
# Ignore blank messages; stop at 'done'.
while True:
    message = input("Message (or 'done'): ")
    if message == "":
        continue                     # nothing typed - skip this pass
    if message == "done":
        break
    print("   sent:", message)

print("Chat closed.")
````

## Definite loops: `for` and `range()`

A **definite loop** runs a known number of times. No initialise/test/update to
get wrong — Python handles all three for you.

```python
for <variable> in range(<how many>):
    <indented body>
```

`range(5)` produces the five numbers `0, 1, 2, 3, 4` — it **starts at 0** and
**stops before 5**. That is the single most common source of off-by-one bugs in
this course.

Rule of thumb: `while` when you do not know how many passes; `for` when you do.

````python
# Print the 7 times table.
for multiplier in range(1, 11):
    print("7 x", multiplier, "=", 7 * multiplier)
````

````python
# range(start, stop, step): a negative step counts DOWN.
for count in range(5, 0, -1):
    print(count)
print("Go!")

# An empty range: start equals stop, so the body never runs at all.
for count in range(3, 3):
    print("you will never see this")
print("finished")
````

`range(start, stop, step)` — `stop` is **always excluded**:

| Written | Produces |
| :--- | :--- |
| `range(4)` | 0, 1, 2, 3 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 20, 5)` | 0, 5, 10, 15 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |
| `range(3, 3)` | *nothing* — the body never runs |

Read `range(a, b)` as *"from `a` up to but not including `b`"* — so
`range(1, 11)` gives 1 through 10, ten numbers.

## The four loop patterns

This is the most valuable half-hour of the course so far. Almost every loop you
write for the rest of the term is one of these four, or a combination:

| Pattern | Initialise before the loop | Update inside the loop |
| :--- | :--- | :--- |
| **Counting** | `count = 0` | `count = count + 1` |
| **Summing** | `total = 0` | `total = total + value` |
| **Maximum** | `largest = None` | `if largest is None or value > largest:` |
| **Minimum** | `smallest = None` | `if smallest is None or value < smallest:` |

Every one has the same shape: **initialise an accumulator, update it once per
item, report it after the loop.**

## Pattern 1 — Counting

*"How many things were there?"*

- Before the loop: `count = 0`
- Inside the loop, once per item: `count = count + 1`
- After the loop: report `count`

Counting only ever cares that an item *happened*, never what its value was.
Put the update inside an `if` and you get **conditional counting** —
"how many days were above 20?"

````python
use_canned("18", "23", "9", "31", "done")
````

````python
# How many temperatures were entered, and how many were warm (> 20)?
count = 0
warm_days = 0

while True:
    line = input("Temperature (or 'done'): ")
    if line == "done":
        break
    temperature = int(line)
    count = count + 1
    if temperature > 20:
        warm_days = warm_days + 1

print("readings:", count, " warm days:", warm_days)
````

**Trace it** for the readings 18, 23, 9, 31:

| Pass | `temperature` | `count` after | warm? | `warm_days` after |
| :--- | :--- | :--- | :--- | :--- |
| 1 | 18 | 1 | no | 0 |
| 2 | 23 | 2 | yes | 1 |
| 3 | 9 | 3 | no | 1 |
| 4 | 31 | 4 | yes | 2 |
| 5 | `"done"` | 4 | — | 2 |

The `"done"` pass breaks *before* `count` is updated — which is exactly why the
`break` is the first thing in the body.

## Pattern 2 — Summing

*"What do they add up to?"*

- Before the loop: `total = 0`
- Inside the loop: `total = total + value`
- After the loop: report `total`

Combine **summing** with **counting** and you get an **average**:
`total / count`.

> Guard the division: if `count` is 0 you get `ZeroDivisionError`. Always ask
> "what if there were no items at all?"

````python
use_canned("18", "23", "9", "31", "done")
````

````python
# Total and average temperature.
total = 0
count = 0

while True:
    line = input("Temperature (or 'done'): ")
    if line == "done":
        break
    total = total + int(line)
    count = count + 1

if count == 0:
    print("No readings entered.")
else:
    print("total:", total, " average:", round(total / count, 1))
````

| Pass | value | `total` after | `count` after |
| :--- | :--- | :--- | :--- |
| 1 | 18 | 18 | 1 |
| 2 | 23 | 41 | 2 |
| 3 | 9 | 50 | 3 |
| 4 | 31 | 81 | 4 |

`81 / 4 = 20.25`, rounded to `20.2` (Python rounds half to even).

## Pattern 3 — Maximum, and the `None` sentinel

*"What was the biggest?"*

The tempting start is `largest = 0`. **It is wrong.** If every temperature is
below zero, the answer comes back as `0` — a value that was never in the data.

Instead start with `None`, meaning *"I have not seen anything yet."*

```python
largest = None
...
if largest is None or value > largest:
    largest = value
```

`None` is a value of its own type, `NoneType`. Compare it with `is` / `is not`,
never with `==`.

````python
use_canned("18", "23", "9", "31", "done")
````

````python
# The highest temperature seen so far.
largest = None

while True:
    line = input("Temperature (or 'done'): ")
    if line == "done":
        break
    temperature = int(line)
    if largest is None or temperature > largest:
        largest = temperature

print("largest:", largest)
````

| Pass | `temperature` | `largest` before | condition | `largest` after |
| :--- | :--- | :--- | :--- | :--- |
| 1 | 18 | `None` | `is None` → `True` | 18 |
| 2 | 23 | 18 | `23 > 18` → `True` | 23 |
| 3 | 9 | 23 | `9 > 23` → `False` | 23 |
| 4 | 31 | 23 | `31 > 23` → `True` | 31 |

**Why the order matters:** Python's `or` is *short-circuit*. Because
`largest is None` is tested first, `temperature > largest` is never evaluated
on pass 1 — which is what stops the `TypeError` from comparing `int` to `None`.
Swap the two halves and pass 1 crashes.

### Predict the output (3)

```python
largest = None
for number in range(4, 0, -1):
    if largest is None or number > largest:
        largest = number
    print(number, largest)
print("final:", largest)
```

What are the five lines of output?

**Answer:**

```
4 4
3 4
2 4
1 4
final: 4
```

`range(4, 0, -1)` counts **down**: 4, 3, 2, 1 (0 is excluded). `largest` is set
on the very first pass and never beaten again — the biggest value arrived
first. The pattern still works: it does not care about order.

## Pattern 4 — Minimum

Identical to maximum with two characters changed: `>` becomes `<`, and the
variable is renamed.

```python
smallest = None
...
if smallest is None or value < smallest:
    smallest = value
```

The `None` start matters even more here: `smallest = 0` would report `0` for
any data set of positive numbers, and `smallest = 999999` quietly breaks the
day someone enters a bigger number.

````python
use_canned("18", "23", "9", "31", "done")
````

````python
# Coldest and warmest in one pass - the two patterns side by side.
smallest = None
largest = None

while True:
    line = input("Temperature (or 'done'): ")
    if line == "done":
        break
    temperature = int(line)
    if smallest is None or temperature < smallest:
        smallest = temperature
    if largest is None or temperature > largest:
        largest = temperature

print("coldest:", smallest, " warmest:", largest, " range:", largest - smallest)
````

## Debugging loops

When a loop misbehaves, there are only four questions:

1. **Does it start right?** Print the accumulator *before* the loop.
2. **Does it run the right number of times?** Add a counter and print it after.
3. **Does the body do the right thing on pass 1?** Print every variable at the
   top of the body — then delete the print when it works.
4. **Does it stop?** Name the line that makes the condition become `False`.

The universal tool: **put a `print()` inside the loop showing every variable
that changes.** Read the trace like the tables we have been drawing.

````python
# A debugging print, temporarily added to a summing loop.
total = 0
for number in range(1, 6):
    total = total + number
    print("  number =", number, " total =", total)   # remove when fixed

print("final total:", total)
````

## Live coding 1 — together

We write this from a blank cell, out loud, as a class.

**Problem.** Ask the user for numbers until they type `done`. Ignore any line
that is not a whole number. Then report how many valid numbers were entered,
their total, their average, the largest, and the smallest.

This is all four patterns in one loop. Start by writing the *initialisations*
before you write anything else.

````python
use_canned("7", "banana", "42", "-3", "done")
````

````python
# Live coding: read numbers until 'done'.
# Report count, total, average, largest, smallest. Skip bad input.
````

````python
# Completed solution
count = 0
total = 0
largest = None
smallest = None

while True:
    line = input("Number (or 'done'): ")
    if line == "done":
        break
    try:
        number = int(line)
    except ValueError:
        print("   not a number, ignoring:", line)
        continue
    count = count + 1
    total = total + number
    if largest is None or number > largest:
        largest = number
    if smallest is None or number < smallest:
        smallest = number

if count == 0:
    print("No numbers entered.")
else:
    print("count:", count, " total:", total)
    print("average:", round(total / count, 2))
    print("largest:", largest, " smallest:", smallest)
````

## Live coding 2 — together

**Problem.** Write a function `count_multiples(limit, factor)` that returns how
many whole numbers from 1 up to and including `limit` are multiples of
`factor`. Then use it to answer: how many numbers from 1 to 100 are multiples
of 7?

Two things to get right: the `range()` bounds, and the counting pattern inside
a function (Week 4 + Week 5 together).

````python
# Live coding: count_multiples(limit, factor)
# How many numbers in 1..limit divide evenly by factor?
````

````python
# Completed solution
def count_multiples(limit, factor):
    """Return how many numbers in 1..limit are multiples of factor."""
    count = 0
    for number in range(1, limit + 1):
        if number % factor == 0:
            count = count + 1
    return count


print("multiples of 7 up to 100:", count_multiples(100, 7))
print("multiples of 25 up to 100:", count_multiples(100, 25))
print("multiples of 3 up to 10: ", count_multiples(10, 3))
````

## Common errors this week

**1. `for i in 5:` — you cannot loop over a number.**

```
TypeError: 'int' object is not iterable
```
*Meaning:* `for` needs something that produces items. Write `range(5)`.

**2. Forgetting the update line → the loop never ends.**
No error message at all. The cell shows `[*]` forever. *Fix:* interrupt the
kernel, then find the variable in your condition and make sure the body changes
it.

````python
# Error 1, on purpose.
for i in 5:
    print(i)
````

**3. Off-by-one from `range()`.**
`range(1, 10)` gives 1–9, not 1–10. To include `n`, write `range(1, n + 1)`.
*Symptom:* your total or count is short by exactly one item. No error message —
this is a **logic** error, and only tracing finds it.

**4. Comparing to `None` with `==`, or putting the test second.**

```python
if value > largest or largest is None:      # WRONG order
```
```
TypeError: '>' not supported between instances of 'int' and 'NoneType'
```
*Meaning:* the first pass compared a number to `None`. The `is None` test must
come **first** so `or` short-circuits.

## Summary

- **Updating a variable** (`x = x + 1`) is the engine of every loop; initialise
  before you update.
- **`while`** is the indefinite loop: initialise, test, **update**. If nothing
  changes the condition, it never stops.
- **`for` + `range()`** is the definite loop. `range(a, b)` stops *before* `b`.
- **`break`** leaves the loop; **`continue`** skips to the next pass.
  `while True:` + `break` is the loop-and-a-half.
- **Four patterns:** counting (`count = 0`), summing (`total = 0`), maximum and
  minimum (start at `None`, test `is None` **first**).
